# DATASET LOADING

```python
import os
import pandas as pd
from collections import Counter

# ==============================================================
# STEP 1: DATASET COLLECTION AND LOADING
# ==============================================================

print("=" * 100)
print("STEP 1: MUSIC EMOTION DATASET COLLECTION AND LOADING")
print("=" * 100)

# --------------------------------------------------------------
# 1. DEFINE DATASET PATHS
# --------------------------------------------------------------

audio_path = r"F:\code\emotion_expression\Dataset\Audios"
image_path = r"F:\code\emotion_expression\Dataset\Images"
csv_path = r"F:\code\emotion_expression\Dataset\music_emotion_dataset.csv"

# --------------------------------------------------------------
# 2. VERIFY DATASET PATHS
# --------------------------------------------------------------

print("\n[1] VERIFYING DATASET PATHS")
print("-" * 100)

paths = {
    "Audio Directory": audio_path,
    "Image Directory": image_path,
    "CSV Dataset": csv_path
}

for name, path in paths.items():
    if os.path.exists(path):
        print(f"{name}: FOUND")
        print(f"Path: {path}")
    else:
        print(f"{name}: NOT FOUND")
        print(f"Path: {path}")

# Stop if any required path is missing
if not all(os.path.exists(path) for path in paths.values()):
    raise FileNotFoundError(
        "One or more dataset paths are invalid. Please verify the paths."
    )

# --------------------------------------------------------------
# 3. LOAD CSV DATASET
# --------------------------------------------------------------

print("\n[2] LOADING CSV DATASET")
print("-" * 100)

try:
    df = pd.read_csv(csv_path, encoding="utf-8")
except UnicodeDecodeError:
    df = pd.read_csv(csv_path, encoding="latin1")

print("CSV dataset loaded successfully.")
print(f"Dataset Shape: {df.shape}")
print(f"Total Records: {df.shape[0]}")
print(f"Total Attributes: {df.shape[1]}")

# --------------------------------------------------------------
# 4. DISPLAY COLUMN INFORMATION
# --------------------------------------------------------------

print("\n[3] DATASET COLUMN INFORMATION")
print("-" * 100)

for index, column in enumerate(df.columns, start=1):
    print(f"{index:02d}. {column} ({df[column].dtype})")

# --------------------------------------------------------------
# 5. DATASET PREVIEW
# --------------------------------------------------------------

print("\n[4] FIRST FIVE RECORDS")
print("-" * 100)

print(df.head())

# --------------------------------------------------------------
# 6. MISSING VALUE ANALYSIS
# --------------------------------------------------------------

print("\n[5] MISSING VALUE ANALYSIS")
print("-" * 100)

missing_values = df.isnull().sum()

print(f"Total Missing Values: {missing_values.sum()}")

missing_columns = missing_values[missing_values > 0]

if len(missing_columns) > 0:
    print("\nColumns containing missing values:")
    print(missing_columns.to_string())
else:
    print("No missing values found.")

# --------------------------------------------------------------
# 7. DUPLICATE RECORD ANALYSIS
# --------------------------------------------------------------

print("\n[6] DUPLICATE RECORD ANALYSIS")
print("-" * 100)

duplicate_count = df.duplicated().sum()

print(f"Total Duplicate Records: {duplicate_count}")

# --------------------------------------------------------------
# 8. TARGET CLASS DISTRIBUTION
# --------------------------------------------------------------

print("\n[7] EMOTION CLASS DISTRIBUTION")
print("-" * 100)

target_column = "Emotion_Class"

if target_column in df.columns:

    emotion_distribution = df[target_column].value_counts(dropna=False)

    print("Emotion Class Counts:")
    print(emotion_distribution.to_string())

    print("\nEmotion Class Percentages:")

    emotion_percentage = (
        df[target_column]
        .value_counts(normalize=True, dropna=False) * 100
    )

    for emotion, percentage in emotion_percentage.items():
        print(f"{emotion}: {percentage:.2f}%")

    print(f"\nTotal Emotion Classes: {df[target_column].nunique(dropna=True)}")

else:
    print(f"Target column '{target_column}' was not found.")
    print("Available columns:", df.columns.tolist())

# --------------------------------------------------------------
# 9. AUDIO FILE INSPECTION
# --------------------------------------------------------------

print("\n[8] AUDIO DATASET INSPECTION")
print("-" * 100)

audio_extensions = (
    ".wav", ".mp3", ".flac", ".ogg", ".m4a", ".aac", ".wma"
)

audio_files = []

for root, dirs, files in os.walk(audio_path):
    for file in files:
        if file.lower().endswith(audio_extensions):
            audio_files.append(os.path.join(root, file))

audio_extension_counts = Counter(
    os.path.splitext(file)[1].lower()
    for file in audio_files
)

print(f"Total Audio Files: {len(audio_files)}")

print("\nAudio File Format Distribution:")

for extension, count in sorted(audio_extension_counts.items()):
    print(f"{extension}: {count}")

print("\nSample Audio Files:")

for file in audio_files[:10]:
    print(os.path.relpath(file, audio_path))

# --------------------------------------------------------------
# 10. IMAGE FILE INSPECTION
# --------------------------------------------------------------

print("\n[9] IMAGE DATASET INSPECTION")
print("-" * 100)

image_extensions = (
    ".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".webp"
)

image_files = []

for root, dirs, files in os.walk(image_path):
    for file in files:
        if file.lower().endswith(image_extensions):
            image_files.append(os.path.join(root, file))

image_extension_counts = Counter(
    os.path.splitext(file)[1].lower()
    for file in image_files
)

print(f"Total Image Files: {len(image_files)}")

print("\nImage File Format Distribution:")

for extension, count in sorted(image_extension_counts.items()):
    print(f"{extension}: {count}")

print("\nSample Image Files:")

for file in image_files[:10]:
    print(os.path.relpath(file, image_path))

# --------------------------------------------------------------
# 11. DATASET SUMMARY
# --------------------------------------------------------------

print("\n" + "=" * 100)
print("STEP 1: DATASET COLLECTION SUMMARY")
print("=" * 100)

print(f"CSV Dataset Records       : {len(df)}")
print(f"CSV Dataset Attributes    : {len(df.columns)}")
print(f"Total Audio Files         : {len(audio_files)}")
print(f"Total Image Files         : {len(image_files)}")
print(f"Missing Values            : {df.isnull().sum().sum()}")
print(f"Duplicate Records         : {duplicate_count}")

if target_column in df.columns:
    print(f"Emotion Classes           : {df[target_column].nunique(dropna=True)}")

print("\nDataset collection and loading completed successfully.")
print("=" * 100)
```

# PREPROCESSING

```python

import os
import cv2
import numpy as np
import pandas as pd
import soundfile as sf
import librosa
import shutil
import warnings

from scipy.ndimage import gaussian_filter1d
from sklearn.preprocessing import MinMaxScaler
from tqdm import tqdm

warnings.filterwarnings("ignore")

# ==============================================================
# STEP 1: CONFIGURATION
# ==============================================================

print("=" * 100)
print("STEP 2: MULTIMODAL MUSIC EMOTION DATASET PREPROCESSING")
print("=" * 100)

# Input paths
BASE_PATH = r"F:\code\emotion_expression\Dataset"

CSV_PATH = os.path.join(
    BASE_PATH, "music_emotion_dataset.csv"
)

IMAGE_ROOT = os.path.join(
    BASE_PATH, "Images"
)

AUDIO_ROOT = os.path.join(
    BASE_PATH, "Audios"
)

# Output paths
OUTPUT_ROOT = r"F:\code\emotion_expression\Step2_Preprocessed"

OUTPUT_IMAGE_ROOT = os.path.join(
    OUTPUT_ROOT, "Images"
)

OUTPUT_AUDIO_ROOT = os.path.join(
    OUTPUT_ROOT, "Audios"
)

OUTPUT_CSV = os.path.join(
    OUTPUT_ROOT, "music_emotion_dataset_preprocessed.csv"
)

# Image configuration
IMAGE_SIZE = (224, 224)

# Gaussian filtering configuration
GAUSSIAN_SIGMA = 1.0

# GSR column
GSR_COLUMN = "Skin_Conductivity"

# ==============================================================
# STEP 2: VERIFY INPUT PATHS
# ==============================================================

print("\n[1] VERIFYING DATASET PATHS")
print("-" * 100)

for path in [CSV_PATH, IMAGE_ROOT, AUDIO_ROOT]:

    if not os.path.exists(path):
        raise FileNotFoundError(
            f"Required dataset path not found: {path}"
        )

    print(f"FOUND: {path}")

os.makedirs(OUTPUT_IMAGE_ROOT, exist_ok=True)
os.makedirs(OUTPUT_AUDIO_ROOT, exist_ok=True)

# ==============================================================
# STEP 3: LOAD CSV DATASET
# ==============================================================

print("\n[2] LOADING CSV DATASET")
print("-" * 100)

df = pd.read_csv(CSV_PATH)

print("Original Dataset Shape:", df.shape)
print("Total Records:", len(df))

required_columns = [
    "Emotion_Class",
    "Image_Path",
    "Audio_Path",
    "Image_ID",
    "Audio_ID",
    GSR_COLUMN
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required CSV columns: {missing_columns}"
    )

# ==============================================================
# STEP 4: IMAGE PREPROCESSING - IMAGE RESIZING
# ==============================================================

print("\n[3] IMAGE PREPROCESSING - IMAGE RESIZING")
print("-" * 100)

image_success = 0
image_failed = 0

processed_image_paths = []

for idx, row in tqdm(
    df.iterrows(),
    total=len(df),
    desc="Resizing Images"
):

    source_path = str(row["Image_Path"])

    emotion = str(row["Emotion_Class"])

    image_id = str(row["Image_ID"])

    if not os.path.isfile(source_path):
        image_failed += 1
        processed_image_paths.append("")
        continue

    try:

        image = cv2.imread(source_path)

        if image is None:
            image_failed += 1
            processed_image_paths.append("")
            continue

        # Resize image to 224 x 224
        resized_image = cv2.resize(
            image,
            IMAGE_SIZE,
            interpolation=cv2.INTER_AREA
        )

        # Save into corresponding emotion folder
        emotion_folder = os.path.join(
            OUTPUT_IMAGE_ROOT, emotion
        )

        os.makedirs(emotion_folder, exist_ok=True)

        output_path = os.path.join(
            emotion_folder,
            image_id + ".jpg"
        )

        cv2.imwrite(
            output_path,
            resized_image,
            [cv2.IMWRITE_JPEG_QUALITY, 95]
        )

        processed_image_paths.append(output_path)

        image_success += 1

    except Exception as e:

        image_failed += 1
        processed_image_paths.append("")

        print(
            f"\nImage Error [{image_id}]: {e}"
        )

df["Image_Path"] = processed_image_paths

print("\nIMAGE PREPROCESSING SUMMARY")
print("-" * 60)
print(f"Successfully Resized Images : {image_success}")
print(f"Failed Images               : {image_failed}")
print(f"Target Image Dimensions     : {IMAGE_SIZE[0]} x {IMAGE_SIZE[1]}")

# ==============================================================
# STEP 5: AUDIO PREPROCESSING - GAUSSIAN FILTERING
# ==============================================================

print("\n[4] AUDIO PREPROCESSING - GAUSSIAN FILTERING")
print("-" * 100)

audio_success = 0
audio_failed = 0

processed_audio_paths = []

for idx, row in tqdm(
    df.iterrows(),
    total=len(df),
    desc="Filtering Audio"
):

    source_path = str(row["Audio_Path"])

    emotion = str(row["Emotion_Class"])

    audio_id = str(row["Audio_ID"])

    if not os.path.isfile(source_path):
        audio_failed += 1
        processed_audio_paths.append("")
        continue

    try:

        # Load audio while preserving its original sample rate
        audio_signal, sample_rate = sf.read(
            source_path,
            dtype="float32",
            always_2d=False
        )

        # Apply Gaussian filtering
        if audio_signal.ndim == 1:

            filtered_audio = gaussian_filter1d(
                audio_signal,
                sigma=GAUSSIAN_SIGMA
            )

        elif audio_signal.ndim == 2:

            filtered_audio = gaussian_filter1d(
                audio_signal,
                sigma=GAUSSIAN_SIGMA,
                axis=0
            )

        else:
            raise ValueError(
                "Unsupported audio signal dimensions"
            )

        # Prevent clipping
        filtered_audio = np.clip(
            filtered_audio,
            -1.0,
            1.0
        )

        # Save filtered audio
        emotion_folder = os.path.join(
            OUTPUT_AUDIO_ROOT, emotion
        )

        os.makedirs(emotion_folder, exist_ok=True)

        output_path = os.path.join(
            emotion_folder,
            audio_id + ".wav"
        )

        sf.write(
            output_path,
            filtered_audio,
            sample_rate,
            subtype="PCM_16"
        )

        processed_audio_paths.append(output_path)

        audio_success += 1

    except Exception as e:

        audio_failed += 1
        processed_audio_paths.append("")

        print(
            f"\nAudio Error [{audio_id}]: {e}"
        )

df["Audio_Path"] = processed_audio_paths

print("\nAUDIO PREPROCESSING SUMMARY")
print("-" * 60)
print(f"Successfully Filtered Audios : {audio_success}")
print(f"Failed Audios                : {audio_failed}")
print(f"Gaussian Filter Sigma        : {GAUSSIAN_SIGMA}")

# ==============================================================
# STEP 6: CSV PREPROCESSING - GSR NORMALIZATION
# ==============================================================

print("\n[5] CSV PREPROCESSING - GSR NORMALIZATION")
print("-" * 100)

print("GSR Column:", GSR_COLUMN)

# Convert GSR values into numeric format
df[GSR_COLUMN] = pd.to_numeric(
    df[GSR_COLUMN],
    errors="coerce"
)

# Check missing values in GSR
gsr_missing = df[GSR_COLUMN].isna().sum()

if gsr_missing > 0:
    raise ValueError(
        f"GSR column contains {gsr_missing} missing or non-numeric values. "
        "Please resolve them before normalization."
    )

# Preserve original values
df["Skin_Conductivity_Original"] = df[GSR_COLUMN].copy()

# Apply Min-Max normalization
scaler = MinMaxScaler()

df[GSR_COLUMN] = scaler.fit_transform(
    df[[GSR_COLUMN]]
).flatten()

print("\nGSR NORMALIZATION SUMMARY")
print("-" * 60)

print(
    "Original GSR Minimum :",
    df["Skin_Conductivity_Original"].min()
)

print(
    "Original GSR Maximum :",
    df["Skin_Conductivity_Original"].max()
)

print(
    "Normalized GSR Minimum :",
    df[GSR_COLUMN].min()
)

print(
    "Normalized GSR Maximum :",
    df[GSR_COLUMN].max()
)

# ==============================================================
# STEP 7: UPDATE PREPROCESSED MEDIA PATHS
# ==============================================================

print("\n[6] UPDATING PREPROCESSED MEDIA PATHS")
print("-" * 100)

df["Image_Path"] = df["Image_Path"].replace("", np.nan)
df["Audio_Path"] = df["Audio_Path"].replace("", np.nan)

print(
    "Valid Image Paths:",
    df["Image_Path"].notna().sum()
)

print(
    "Valid Audio Paths:",
    df["Audio_Path"].notna().sum()
)

# ==============================================================
# STEP 8: SAVE PREPROCESSED CSV
# ==============================================================

print("\n[7] SAVING PREPROCESSED CSV DATASET")
print("-" * 100)

df.to_csv(
    OUTPUT_CSV,
    index=False
)

print("Preprocessed CSV Saved Successfully:")
print(OUTPUT_CSV)

# ==============================================================
# STEP 9: FINAL DATASET SUMMARY
# ==============================================================

print("\n" + "=" * 100)
print("STEP 2: MULTIMODAL DATASET PREPROCESSING SUMMARY")
print("=" * 100)

print(f"{'Original CSV Records':35}: {len(df)}")
print(f"{'Preprocessed Image Count':35}: {image_success}")
print(f"{'Failed Image Count':35}: {image_failed}")
print(f"{'Preprocessed Audio Count':35}: {audio_success}")
print(f"{'Failed Audio Count':35}: {audio_failed}")
print(f"{'Image Dimensions':35}: {IMAGE_SIZE}")
print(f"{'Gaussian Filter Sigma':35}: {GAUSSIAN_SIGMA}")
print(f"{'GSR Normalization':35}: Min-Max Normalization")
print(f"{'GSR Minimum After Normalization':35}: {df[GSR_COLUMN].min():.4f}")
print(f"{'GSR Maximum After Normalization':35}: {df[GSR_COLUMN].max():.4f}")
print(f"{'Final CSV Shape':35}: {df.shape}")

print("\nEmotion Class Distribution:")
print(df["Emotion_Class"].value_counts())

print("\nOutput Directory:")
print(OUTPUT_ROOT)

print("\nPreprocessed CSV:")
print(OUTPUT_CSV)

print("\n" + "=" * 100)
print("STEP 2: MULTIMODAL DATASET PREPROCESSING COMPLETED")
print("=" * 100)
```

# FEATURE EXTRACTION

```python

import os
import cv2
import numpy as np
import pandas as pd
import librosa
import warnings

from skimage.feature import hog
from tqdm import tqdm

warnings.filterwarnings("ignore")

# ==============================================================
# STEP 1: CONFIGURATION
# ==============================================================

print("=" * 100)
print("STEP 3: MULTIMODAL FEATURE EXTRACTION")
print("=" * 100)

# Step 2 input directory
INPUT_ROOT = r"F:\code\emotion_expression\Step2_Preprocessed"

CSV_PATH = os.path.join(
    INPUT_ROOT,
    "music_emotion_dataset_preprocessed.csv"
)

# Output directory
OUTPUT_ROOT = r"F:\code\emotion_expression\Step3_Feature_Extraction"

os.makedirs(OUTPUT_ROOT, exist_ok=True)

# Feature output files
IMAGE_OUTPUT = os.path.join(
    OUTPUT_ROOT, "Image_HOG_Features.csv"
)

AUDIO_OUTPUT = os.path.join(
    OUTPUT_ROOT, "Audio_MFCC_Features.csv"
)

PHYSIO_OUTPUT = os.path.join(
    OUTPUT_ROOT, "Physiological_Features.csv"
)

# HOG configuration
IMAGE_SIZE = (224, 224)
HOG_ORIENTATIONS = 9
PIXELS_PER_CELL = (16, 16)
CELLS_PER_BLOCK = (2, 2)

# MFCC configuration
N_MFCC = 13
N_FFT = 2048
HOP_LENGTH = 512

# ==============================================================
# STEP 2: LOAD PREPROCESSED DATASET
# ==============================================================

print("\n[1] LOADING PREPROCESSED DATASET")
print("-" * 100)

if not os.path.isfile(CSV_PATH):
    raise FileNotFoundError(
        f"Preprocessed CSV not found: {CSV_PATH}"
    )

df = pd.read_csv(CSV_PATH)

print("Dataset Shape:", df.shape)
print("Total Records:", len(df))

required_columns = [
    "Performance_ID",
    "Image_ID",
    "Audio_ID",
    "Image_Path",
    "Audio_Path",
    "Emotion_Class"
]

missing = [
    col for col in required_columns
    if col not in df.columns
]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}"
    )

# ==============================================================
# STEP 3: IMAGE FEATURE EXTRACTION - HOG
# ==============================================================

print("\n[2] IMAGE FEATURE EXTRACTION - HOG")
print("-" * 100)

image_feature_records = []
image_failed = 0

for _, row in tqdm(
    df.iterrows(),
    total=len(df),
    desc="Extracting HOG Features"
):

    image_path = str(row["Image_Path"])

    try:

        if not os.path.isfile(image_path):
            raise FileNotFoundError(image_path)

        # Read image
        image = cv2.imread(
            image_path,
            cv2.IMREAD_GRAYSCALE
        )

        if image is None:
            raise ValueError("Unable to read image")

        # Ensure consistent image dimensions
        image = cv2.resize(
            image,
            IMAGE_SIZE,
            interpolation=cv2.INTER_AREA
        )

        # Normalize image intensity
        image = image.astype(np.float32) / 255.0

        # Extract HOG descriptor
        hog_features = hog(
            image,
            orientations=HOG_ORIENTATIONS,
            pixels_per_cell=PIXELS_PER_CELL,
            cells_per_block=CELLS_PER_BLOCK,
            block_norm="L2-Hys",
            visualize=False,
            feature_vector=True
        )

        record = {
            "Performance_ID": row["Performance_ID"],
            "Image_ID": row["Image_ID"],
            "Emotion_Class": row["Emotion_Class"]
        }

        # Store each HOG feature
        for i, value in enumerate(hog_features):

            record[f"HOG_{i+1}"] = float(value)

        image_feature_records.append(record)

    except Exception as e:

        image_failed += 1

        print(
            f"\nImage Error - {row['Image_ID']}: {e}"
        )

# Convert extracted features to DataFrame
image_features_df = pd.DataFrame(
    image_feature_records
)

# Save HOG features
image_features_df.to_csv(
    IMAGE_OUTPUT,
    index=False
)

print("\nIMAGE HOG EXTRACTION SUMMARY")
print("-" * 70)
print("Successfully Processed Images:", len(image_features_df))
print("Failed Images:", image_failed)

if len(image_features_df) > 0:
    print(
        "HOG Features per Image:",
        image_features_df.shape[1] - 3
    )

print("HOG Dataset Shape:", image_features_df.shape)
print("Saved:", IMAGE_OUTPUT)

# ==============================================================
# STEP 4: AUDIO FEATURE EXTRACTION - MFCC
# ==============================================================

print("\n[3] AUDIO FEATURE EXTRACTION - MFCC")
print("-" * 100)

audio_feature_records = []
audio_failed = 0

for _, row in tqdm(
    df.iterrows(),
    total=len(df),
    desc="Extracting MFCC Features"
):

    audio_path = str(row["Audio_Path"])

    try:

        if not os.path.isfile(audio_path):
            raise FileNotFoundError(audio_path)

        # Load audio while preserving original sample rate
        audio_signal, sample_rate = librosa.load(
            audio_path,
            sr=None,
            mono=True
        )

        if len(audio_signal) == 0:
            raise ValueError("Empty audio signal")

        # Extract MFCC coefficients
        mfcc = librosa.feature.mfcc(
            y=audio_signal,
            sr=sample_rate,
            n_mfcc=N_MFCC,
            n_fft=N_FFT,
            hop_length=HOP_LENGTH
        )

        # Calculate statistical descriptors
        mfcc_mean = np.mean(mfcc, axis=1)
        mfcc_std = np.std(mfcc, axis=1)
        mfcc_median = np.median(mfcc, axis=1)

        record = {
            "Performance_ID": row["Performance_ID"],
            "Audio_ID": row["Audio_ID"],
            "Emotion_Class": row["Emotion_Class"],
            "Sample_Rate": sample_rate,
            "Audio_Duration_sec": len(audio_signal) / sample_rate
        }

        # MFCC mean
        for i in range(N_MFCC):

            record[f"MFCC_Mean_{i+1}"] = float(
                mfcc_mean[i]
            )

        # MFCC standard deviation
        for i in range(N_MFCC):

            record[f"MFCC_Std_{i+1}"] = float(
                mfcc_std[i]
            )

        # MFCC median
        for i in range(N_MFCC):

            record[f"MFCC_Median_{i+1}"] = float(
                mfcc_median[i]
            )

        audio_feature_records.append(record)

    except Exception as e:

        audio_failed += 1

        print(
            f"\nAudio Error - {row['Audio_ID']}: {e}"
        )

# Convert to DataFrame
audio_features_df = pd.DataFrame(
    audio_feature_records
)

# Save MFCC features
audio_features_df.to_csv(
    AUDIO_OUTPUT,
    index=False
)

print("\nAUDIO MFCC EXTRACTION SUMMARY")
print("-" * 70)
print("Successfully Processed Audios:", len(audio_features_df))
print("Failed Audios:", audio_failed)

if len(audio_features_df) > 0:
    print(
        "MFCC Features per Audio:",
        audio_features_df.shape[1] - 5
    )

print("MFCC Dataset Shape:", audio_features_df.shape)
print("Saved:", AUDIO_OUTPUT)

# ==============================================================
# STEP 5: PHYSIOLOGICAL FEATURE EXTRACTION
# PPG-RELATED HEART RATE, HRV AND GSR
# ==============================================================

print("\n[4] PHYSIOLOGICAL FEATURE EXTRACTION")
print("-" * 100)

# Available physiological columns
physiological_columns = [
    "Heart_Rate_BPM",
    "HRV_Score",
    "Skin_Conductivity"
]

missing_physiological = [
    col for col in physiological_columns
    if col not in df.columns
]

if missing_physiological:
    raise ValueError(
        "Missing physiological columns: "
        + str(missing_physiological)
    )

# Check for raw PPG waveform
ppg_candidates = [
    col for col in df.columns
    if "ppg" in col.lower()
]

if len(ppg_candidates) == 0:

    print(
        "Raw PPG waveform column not found."
    )

    print(
        "Heart_Rate_BPM will be retained as a "
        "PPG-related heart-rate indicator, not as "
        "a waveform-derived PPG feature."
    )

physiological_records = []

for _, row in tqdm(
    df.iterrows(),
    total=len(df),
    desc="Extracting Physiological Features"
):

    record = {
        "Performance_ID": row["Performance_ID"],
        "Emotion_Class": row["Emotion_Class"],

        # PPG-related heart-rate indicator
        "Heart_Rate_BPM": row["Heart_Rate_BPM"],

        # HRV feature
        "HRV_Score": row["HRV_Score"],

        # Normalized GSR feature
        "GSR_Normalized": row["Skin_Conductivity"]
    }

    physiological_records.append(record)

physiological_df = pd.DataFrame(
    physiological_records
)

# Save physiological features
physiological_df.to_csv(
    PHYSIO_OUTPUT,
    index=False
)

print("\nPHYSIOLOGICAL FEATURE SUMMARY")
print("-" * 70)

print(
    "Successfully Processed Records:",
    len(physiological_df)
)

print(
    "PPG-Related Feature: Heart_Rate_BPM"
)

print(
    "HRV Feature: HRV_Score"
)

print(
    "GSR Feature: Skin_Conductivity"
)

print(
    "Physiological Dataset Shape:",
    physiological_df.shape
)

print("Saved:", PHYSIO_OUTPUT)

# ==============================================================
# STEP 6: FEATURE EXTRACTION VALIDATION
# ==============================================================

print("\n[5] FEATURE EXTRACTION VALIDATION")
print("-" * 100)

print("\nIMAGE HOG FEATURES")
print("-" * 60)

if not image_features_df.empty:

    print("Dataset Shape:", image_features_df.shape)
    print(
        "Missing Values:",
        image_features_df.isnull().sum().sum()
    )

print("\nAUDIO MFCC FEATURES")
print("-" * 60)

if not audio_features_df.empty:

    print("Dataset Shape:", audio_features_df.shape)
    print(
        "Missing Values:",
        audio_features_df.isnull().sum().sum()
    )

print("\nPHYSIOLOGICAL FEATURES")
print("-" * 60)

print("Dataset Shape:", physiological_df.shape)

print(
    "Missing Values:",
    physiological_df.isnull().sum().sum()
)

# ==============================================================
# STEP 7: FINAL SUMMARY
# ==============================================================

print("\n" + "=" * 100)
print("STEP 3: MULTIMODAL FEATURE EXTRACTION SUMMARY")
print("=" * 100)

print(
    f"{'Original Dataset Records':40}: {len(df)}"
)

print(
    f"{'HOG Processed Images':40}: {len(image_features_df)}"
)

print(
    f"{'HOG Feature Count':40}: "
    f"{image_features_df.shape[1] - 3 if not image_features_df.empty else 0}"
)

print(
    f"{'MFCC Processed Audios':40}: {len(audio_features_df)}"
)

print(
    f"{'MFCC Feature Count':40}: "
    f"{audio_features_df.shape[1] - 5 if not audio_features_df.empty else 0}"
)

print(
    f"{'Physiological Records':40}: {len(physiological_df)}"
)

print(
    f"{'Physiological Feature Count':40}: "
    f"{physiological_df.shape[1] - 2}"
)

print("\nGENERATED FEATURE FILES")
print("-" * 70)

print("1. Image HOG Features:")
print(IMAGE_OUTPUT)

print("\n2. Audio MFCC Features:")
print(AUDIO_OUTPUT)

print("\n3. Physiological Features:")
print(PHYSIO_OUTPUT)

print("\nOUTPUT DIRECTORY:")
print(OUTPUT_ROOT)

print("\n" + "=" * 100)
print("STEP 3: MULTIMODAL FEATURE EXTRACTION COMPLETED")
print("=" * 100)
```

```python

import os
import gc
import time
import warnings
import multiprocessing
import numpy as np
import pandas as pd
import cv2
import librosa
import soundfile as sf

from skimage.feature import hog
from scipy.signal import resample_poly
from joblib import Parallel, delayed
from tqdm import tqdm

warnings.filterwarnings("ignore")

# ============================================================
# STEP 1: CONFIGURATION
# ============================================================

INPUT_CSV = (
    r"F:\code\emotion_expression\Step2_Preprocessed"
    r"\music_emotion_dataset_preprocessed.csv"
)

OUTPUT_DIR = r"F:\code\emotion_expression\Step3_Feature_Extraction"

os.makedirs(OUTPUT_DIR, exist_ok=True)

HOG_OUTPUT = os.path.join(OUTPUT_DIR, "Image_HOG_Features.csv")
MFCC_OUTPUT = os.path.join(OUTPUT_DIR, "Audio_MFCC_Features.csv")
PHYSIO_OUTPUT = os.path.join(OUTPUT_DIR, "Physiological_Features.csv")

IMAGE_SIZE = (224, 224)

HOG_ORIENTATIONS = 9
PIXELS_PER_CELL = (16, 16)
CELLS_PER_BLOCK = (2, 2)

N_MFCC = 13
N_FFT = 2048
HOP_LENGTH = 512
TARGET_SR = 22050

BATCH_SIZE = 100

# Reserve one CPU core for the operating system.
CPU_COUNT = multiprocessing.cpu_count()
N_JOBS = max(1, CPU_COUNT - 1)

print("=" * 75)
print("FAST PARALLEL MULTIMODAL FEATURE EXTRACTION")
print("=" * 75)

print("Available CPU Cores :", CPU_COUNT)
print("Parallel Workers    :", N_JOBS)
print("Batch Size          :", BATCH_SIZE)

start_time = time.time()

# ============================================================
# STEP 2: LOAD DATASET
# ============================================================

df = pd.read_csv(INPUT_CSV)

print("\nDataset Shape:", df.shape)

required_columns = [
    "Performance_ID",
    "Image_ID",
    "Image_Path",
    "Audio_ID",
    "Audio_Path",
    "Emotion_Class"
]

missing = [c for c in required_columns if c not in df.columns]

if missing:
    raise ValueError(f"Missing required columns: {missing}")

df = df.reset_index(drop=True)

print("\nEmotion Distribution:")
print(df["Emotion_Class"].value_counts())

# ============================================================
# STEP 3: HELPER FUNCTION FOR BATCHES
# ============================================================

def batches(data, batch_size):
    for start in range(0, len(data), batch_size):
        yield data.iloc[start:start + batch_size]


# ============================================================
# STEP 4: PARALLEL HOG EXTRACTION
# ============================================================

def extract_hog(row):

    try:
        image_path = str(row["Image_Path"])

        if not os.path.isfile(image_path):
            return None

        image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

        if image is None:
            return None

        image = cv2.resize(
            image,
            IMAGE_SIZE,
            interpolation=cv2.INTER_AREA
        )

        features = hog(
            image,
            orientations=HOG_ORIENTATIONS,
            pixels_per_cell=PIXELS_PER_CELL,
            cells_per_block=CELLS_PER_BLOCK,
            block_norm="L2-Hys",
            feature_vector=True
        )

        return (
            row["Performance_ID"],
            row["Image_ID"],
            row["Emotion_Class"],
            features.astype(np.float32)
        )

    except Exception:
        return None


print("\n" + "=" * 75)
print("STEP 3A: PARALLEL HOG IMAGE FEATURE EXTRACTION")
print("=" * 75)

hog_columns = [
    "Performance_ID",
    "Image_ID",
    "Emotion_Class"
] + [
    f"HOG_{i+1}" for i in range(6084)
]

hog_success = 0
hog_failed = 0

if os.path.exists(HOG_OUTPUT):
    os.remove(HOG_OUTPUT)

for batch_no, batch_df in enumerate(
    batches(df, BATCH_SIZE), start=1
):

    rows = batch_df.to_dict("records")

    results = Parallel(
        n_jobs=N_JOBS,
        backend="loky",
        batch_size=16
    )(
        delayed(extract_hog)(row)
        for row in rows
    )

    output_rows = []

    for result in results:

        if result is None:
            hog_failed += 1
            continue

        performance_id, image_id, emotion, features = result

        output_rows.append(
            [performance_id, image_id, emotion]
            + features.tolist()
        )

        hog_success += 1

    if output_rows:
        pd.DataFrame(
            output_rows,
            columns=hog_columns
        ).to_csv(
            HOG_OUTPUT,
            mode="a",
            header=not os.path.exists(HOG_OUTPUT)
            or os.path.getsize(HOG_OUTPUT) == 0,
            index=False
        )

    del results, output_rows, rows
    gc.collect()

    print(
        f"HOG Batch {batch_no}: "
        f"Processed {min(batch_no * BATCH_SIZE, len(df))}/{len(df)} "
        f"| Successful: {hog_success} | Failed: {hog_failed}"
    )

print("\nHOG Extraction Completed")
print("Successful Images:", hog_success)
print("Failed Images:", hog_failed)
print("Saved:", HOG_OUTPUT)


# ============================================================
# STEP 5: PARALLEL MFCC AUDIO FEATURE EXTRACTION
# ============================================================

def extract_mfcc(row):

    try:
        audio_path = str(row["Audio_Path"])

        if not os.path.isfile(audio_path):
            return None

        # Read audio without unnecessary librosa file-loading overhead.
        signal, sr = sf.read(
            audio_path,
            dtype="float32",
            always_2d=False
        )

        # Convert stereo to mono.
        if signal.ndim == 2:
            signal = np.mean(signal, axis=1)

        if len(signal) == 0:
            return None

        # Resample only when necessary.
        if sr != TARGET_SR:

            from math import gcd

            divisor = gcd(int(sr), TARGET_SR)

            signal = resample_poly(
                signal,
                TARGET_SR // divisor,
                int(sr) // divisor
            ).astype(np.float32)

            sr = TARGET_SR

        signal = np.asarray(signal, dtype=np.float32)

        # MFCC extraction.
        mfcc = librosa.feature.mfcc(
            y=signal,
            sr=sr,
            n_mfcc=N_MFCC,
            n_fft=N_FFT,
            hop_length=HOP_LENGTH
        )

        mfcc_features = np.concatenate([
            np.mean(mfcc, axis=1),
            np.std(mfcc, axis=1),
            np.median(mfcc, axis=1)
        ])

        duration = len(signal) / sr

        return (
            row["Performance_ID"],
            row["Audio_ID"],
            row["Emotion_Class"],
            sr,
            duration,
            mfcc_features.astype(np.float32)
        )

    except Exception:
        return None


print("\n" + "=" * 75)
print("STEP 3B: PARALLEL MFCC AUDIO FEATURE EXTRACTION")
print("=" * 75)

mfcc_feature_names = (
    [f"MFCC_Mean_{i+1}" for i in range(N_MFCC)]
    + [f"MFCC_Std_{i+1}" for i in range(N_MFCC)]
    + [f"MFCC_Median_{i+1}" for i in range(N_MFCC)]
)

mfcc_columns = [
    "Performance_ID",
    "Audio_ID",
    "Emotion_Class",
    "Sample_Rate",
    "Audio_Duration_sec"
] + mfcc_feature_names

mfcc_success = 0
mfcc_failed = 0

if os.path.exists(MFCC_OUTPUT):
    os.remove(MFCC_OUTPUT)

for batch_no, batch_df in enumerate(
    batches(df, BATCH_SIZE), start=1
):

    rows = batch_df.to_dict("records")

    results = Parallel(
        n_jobs=N_JOBS,
        backend="loky",
        batch_size=8
    )(
        delayed(extract_mfcc)(row)
        for row in rows
    )

    output_rows = []

    for result in results:

        if result is None:
            mfcc_failed += 1
            continue

        performance_id, audio_id, emotion, sr, duration, features = result

        output_rows.append(
            [
                performance_id,
                audio_id,
                emotion,
                sr,
                round(duration, 4)
            ] + features.tolist()
        )

        mfcc_success += 1

    if output_rows:
        pd.DataFrame(
            output_rows,
            columns=mfcc_columns
        ).to_csv(
            MFCC_OUTPUT,
            mode="a",
            header=not os.path.exists(MFCC_OUTPUT)
            or os.path.getsize(MFCC_OUTPUT) == 0,
            index=False
        )

    del results, output_rows, rows
    gc.collect()

    print(
        f"MFCC Batch {batch_no}: "
        f"Processed {min(batch_no * BATCH_SIZE, len(df))}/{len(df)} "
        f"| Successful: {mfcc_success} | Failed: {mfcc_failed}"
    )

print("\nMFCC Extraction Completed")
print("Successful Audios:", mfcc_success)
print("Failed Audios:", mfcc_failed)
print("Saved:", MFCC_OUTPUT)


# ============================================================
# STEP 6: PHYSIOLOGICAL FEATURE EXTRACTION
# ============================================================

print("\n" + "=" * 75)
print("STEP 3C: PHYSIOLOGICAL FEATURE EXTRACTION")
print("=" * 75)

physiological_columns = [
    "Heart_Rate_BPM",
    "HRV_Score",
    "Skin_Conductivity",
    "Respiration_Rate",
    "Body_Temperature_C",
    "Oxygen_Saturation",
    "Stress_Index",
    "Fatigue_Score"
]

available_columns = [
    col for col in physiological_columns
    if col in df.columns
]

if not available_columns:
    raise ValueError(
        "No expected physiological columns found in the input CSV."
    )

print("Available Physiological Features:", available_columns)

physio_df = df[
    ["Performance_ID", "Emotion_Class"] + available_columns
].copy()

# Rename fields to describe their actual meaning.
rename_columns = {
    "Heart_Rate_BPM": "PPG_Related_Heart_Rate_BPM",
    "Skin_Conductivity": "GSR_Normalized"
}

physio_df.rename(
    columns=rename_columns,
    inplace=True
)

# Convert feature columns to numeric values.
for col in physio_df.columns:

    if col not in ["Performance_ID", "Emotion_Class"]:
        physio_df[col] = pd.to_numeric(
            physio_df[col],
            errors="coerce"
        )

physio_df.to_csv(
    PHYSIO_OUTPUT,
    index=False
)

print("\nPhysiological Extraction Completed")
print("Records:", len(physio_df))
print("Features:", len(available_columns))
print("Saved:", PHYSIO_OUTPUT)


# ============================================================
# STEP 7: FINAL VALIDATION
# ============================================================

print("\n" + "=" * 75)
print("STEP 3: FINAL FEATURE EXTRACTION SUMMARY")
print("=" * 75)

def validate_csv(path, expected_columns):

    if not os.path.isfile(path):
        return 0, 0

    result = pd.read_csv(path, nrows=0)

    row_count = sum(
        1 for _ in open(path, encoding="utf-8")
    ) - 1

    return row_count, len(result.columns)


for name, path, expected in [
    ("HOG", HOG_OUTPUT, len(hog_columns)),
    ("MFCC", MFCC_OUTPUT, len(mfcc_columns)),
    ("PHYSIOLOGICAL", PHYSIO_OUTPUT, len(physio_df.columns))
]:

    rows, columns = validate_csv(path, expected)

    print(f"\n{name} FEATURES")
    print("Rows:", rows)
    print("Columns:", columns)
    print("Expected Columns:", expected)
    print("Status:", "SUCCESS" if columns == expected else "CHECK OUTPUT")


print("\n" + "=" * 75)
print("ALL FEATURE EXTRACTION TASKS COMPLETED")
print("=" * 75)

print("HOG Successful:", hog_success)
print("HOG Failed:", hog_failed)

print("MFCC Successful:", mfcc_success)
print("MFCC Failed:", mfcc_failed)

print("Physiological Records:", len(physio_df))

print(
    "\nTotal Execution Time:",
    round(time.time() - start_time, 2),
    "seconds"
)

print("\nOUTPUT DIRECTORY:")
print(OUTPUT_DIR)
```

```python

import os
import gc
import time
import warnings
import pandas as pd
import numpy as np

warnings.filterwarnings("ignore")

# ============================================================
# STEP 1: CONFIGURATION
# ============================================================

BASE_DIR = r"F:\code\emotion_expression"

INPUT_DIR = os.path.join(
    BASE_DIR,
    "Step2_Preprocessed"
)

FEATURE_DIR = os.path.join(
    BASE_DIR,
    "Step3_Feature_Extraction"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "Step4_Feature_Fusion"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

MASTER_FILE = os.path.join(
    INPUT_DIR,
    "music_emotion_dataset_preprocessed.csv"
)

HOG_FILE = os.path.join(
    FEATURE_DIR,
    "Image_HOG_Features.csv"
)

MFCC_FILE = os.path.join(
    FEATURE_DIR,
    "Audio_MFCC_Features.csv"
)

PHYSIO_FILE = os.path.join(
    FEATURE_DIR,
    "Physiological_Features.csv"
)

OUTPUT_FILE = os.path.join(
    OUTPUT_DIR,
    "Multimodal_Fused_Dataset.csv"
)

start_time = time.time()

print("=" * 80)
print("STEP 4: MULTIMODAL FEATURE FUSION")
print("=" * 80)

# ============================================================
# STEP 2: CHECK INPUT FILES
# ============================================================

input_files = [
    MASTER_FILE,
    HOG_FILE,
    MFCC_FILE,
    PHYSIO_FILE
]

for file in input_files:

    if not os.path.isfile(file):
        raise FileNotFoundError(
            f"Required file not found: {file}"
        )

    print("Found:", file)

# ============================================================
# STEP 3: LOAD DATASETS
# ============================================================

print("\nLoading datasets...")

master_df = pd.read_csv(MASTER_FILE)
hog_df = pd.read_csv(HOG_FILE)
mfcc_df = pd.read_csv(MFCC_FILE)
physio_df = pd.read_csv(PHYSIO_FILE)

print("\nDataset Dimensions")
print("-" * 60)

print("Master Dataset       :", master_df.shape)
print("HOG Dataset          :", hog_df.shape)
print("MFCC Dataset         :", mfcc_df.shape)
print("Physiological Dataset:", physio_df.shape)

# ============================================================
# STEP 4: CHECK REQUIRED IDENTIFIERS
# ============================================================

master_keys = [
    "Performance_ID",
    "Image_ID",
    "Audio_ID",
    "Emotion_Class"
]

hog_keys = [
    "Performance_ID",
    "Image_ID",
    "Emotion_Class"
]

mfcc_keys = [
    "Performance_ID",
    "Audio_ID",
    "Emotion_Class"
]

physio_keys = [
    "Performance_ID",
    "Emotion_Class"
]

for name, dataset, columns in [
    ("Master", master_df, master_keys),
    ("HOG", hog_df, hog_keys),
    ("MFCC", mfcc_df, mfcc_keys),
    ("Physiological", physio_df, physio_keys)
]:

    missing = [
        col for col in columns
        if col not in dataset.columns
    ]

    if missing:
        raise ValueError(
            f"{name} dataset missing columns: {missing}"
        )

# ============================================================
# STEP 5: CHECK DUPLICATE IDENTIFIERS
# ============================================================

print("\nChecking duplicate identifiers...")

checks = [
    (
        "Master",
        master_df,
        master_keys
    ),
    (
        "HOG",
        hog_df,
        hog_keys
    ),
    (
        "MFCC",
        mfcc_df,
        mfcc_keys
    ),
    (
        "Physiological",
        physio_df,
        physio_keys
    )
]

for name, dataset, keys in checks:

    duplicate_count = dataset.duplicated(
        subset=keys
    ).sum()

    print(
        f"{name} duplicate records: {duplicate_count}"
    )

    if duplicate_count > 0:
        raise ValueError(
            f"Duplicate identifiers found in {name}. "
            "Resolve duplicates before fusion."
        )

# ============================================================
# STEP 6: SELECT FEATURE COLUMNS
# ============================================================

hog_features = [
    col for col in hog_df.columns
    if col.startswith("HOG_")
]

mfcc_features = [
    col for col in mfcc_df.columns
    if col.startswith("MFCC_")
]

physio_features = [
    col for col in physio_df.columns
    if col not in physio_keys
]

print("\nExtracted Feature Counts")
print("-" * 60)

print("HOG Features          :", len(hog_features))
print("MFCC Features         :", len(mfcc_features))
print("Physiological Features:", len(physio_features))

if len(hog_features) != 6084:
    raise ValueError(
        f"Expected 6084 HOG features, found {len(hog_features)}"
    )

if len(mfcc_features) != 39:
    raise ValueError(
        f"Expected 39 MFCC features, found {len(mfcc_features)}"
    )

if len(physio_features) != 8:
    raise ValueError(
        f"Expected 8 physiological features, found {len(physio_features)}"
    )

# ============================================================
# STEP 7: PREPARE MASTER IDENTIFIERS
# ============================================================

fused_df = master_df[
    master_keys
].copy()

# ============================================================
# STEP 8: MERGE HOG FEATURES
# ============================================================

print("\n" + "=" * 80)
print("MERGING IMAGE HOG FEATURES")
print("=" * 80)

hog_selected = hog_df[
    hog_keys + hog_features
].copy()

fused_df = fused_df.merge(
    hog_selected,
    on=hog_keys,
    how="left",
    validate="one_to_one",
    indicator="_hog_merge"
)

hog_missing = (
    fused_df["_hog_merge"] != "both"
).sum()

print("HOG matched records :", len(fused_df) - hog_missing)
print("HOG unmatched records:", hog_missing)

fused_df.drop(
    columns=["_hog_merge"],
    inplace=True
)

if hog_missing > 0:
    raise ValueError(
        "Some master records do not have matching HOG features."
    )

# ============================================================
# STEP 9: MERGE AUDIO MFCC FEATURES
# ============================================================

print("\n" + "=" * 80)
print("MERGING AUDIO MFCC FEATURES")
print("=" * 80)

mfcc_selected = mfcc_df[
    mfcc_keys + mfcc_features
].copy()

fused_df = fused_df.merge(
    mfcc_selected,
    on=mfcc_keys,
    how="left",
    validate="one_to_one",
    indicator="_mfcc_merge"
)

mfcc_missing = (
    fused_df["_mfcc_merge"] != "both"
).sum()

print("MFCC matched records :", len(fused_df) - mfcc_missing)
print("MFCC unmatched records:", mfcc_missing)

fused_df.drop(
    columns=["_mfcc_merge"],
    inplace=True
)

if mfcc_missing > 0:
    raise ValueError(
        "Some master records do not have matching MFCC features."
    )

# ============================================================
# STEP 10: MERGE PHYSIOLOGICAL FEATURES
# ============================================================

print("\n" + "=" * 80)
print("MERGING PHYSIOLOGICAL FEATURES")
print("=" * 80)

physio_selected = physio_df[
    physio_keys + physio_features
].copy()

fused_df = fused_df.merge(
    physio_selected,
    on=physio_keys,
    how="left",
    validate="one_to_one",
    indicator="_physio_merge"
)

physio_missing = (
    fused_df["_physio_merge"] != "both"
).sum()

print(
    "Physiological matched records:",
    len(fused_df) - physio_missing
)

print(
    "Physiological unmatched records:",
    physio_missing
)

fused_df.drop(
    columns=["_physio_merge"],
    inplace=True
)

if physio_missing > 0:
    raise ValueError(
        "Some master records do not have matching physiological features."
    )

# ============================================================
# STEP 11: ARRANGE FINAL FEATURE COLUMNS
# ============================================================

feature_columns = (
    hog_features
    + mfcc_features
    + physio_features
)

metadata_columns = [
    "Performance_ID",
    "Image_ID",
    "Audio_ID"
]

target_column = "Emotion_Class"

fused_df = fused_df[
    metadata_columns
    + feature_columns
    + [target_column]
].copy()

# ============================================================
# STEP 12: CHECK MISSING VALUES
# ============================================================

print("\n" + "=" * 80)
print("MISSING VALUE ANALYSIS")
print("=" * 80)

missing_total = fused_df.isnull().sum().sum()

print("Total Missing Values:", missing_total)

if missing_total > 0:

    missing_columns = (
        fused_df.isnull()
        .sum()
    )

    print(
        missing_columns[
            missing_columns > 0
        ]
    )

    raise ValueError(
        "Missing values detected after fusion. "
        "Resolve them before modeling."
    )

# ============================================================
# STEP 13: CHECK LABEL DISTRIBUTION
# ============================================================

print("\nEmotion Distribution After Fusion")
print("-" * 60)

print(
    fused_df[target_column].value_counts()
)

# ============================================================
# STEP 14: SAVE FUSED DATASET
# ============================================================

print("\nSaving fused dataset...")

fused_df.to_csv(
    OUTPUT_FILE,
    index=False
)

print("Saved Successfully:", OUTPUT_FILE)

# ============================================================
# STEP 15: FINAL VALIDATION
# ============================================================

print("\n" + "=" * 80)
print("STEP 4: FINAL FUSION SUMMARY")
print("=" * 80)

print("Master Records          :", len(master_df))
print("Fused Records           :", len(fused_df))

print("HOG Features            :", len(hog_features))
print("MFCC Features           :", len(mfcc_features))
print("Physiological Features  :", len(physio_features))

print(
    "Total Features          :",
    len(feature_columns)
)

print(
    "Total Metadata Columns  :",
    len(metadata_columns)
)

print(
    "Total Dataset Columns   :",
    len(fused_df.columns)
)

print("Missing Values          :", missing_total)
print(
    "Duplicate Fused Records :",
    fused_df.duplicated(subset=master_keys).sum()
)

print("\nFinal Dataset Shape:", fused_df.shape)

print("\nFinal CSV:")
print(OUTPUT_FILE)


print("\n" + "=" * 80)
print("STEP 4 MULTIMODAL FEATURE FUSION COMPLETED")
print("=" * 80)
```

```python

import os
import gc
import json
import time
import warnings
import numpy as np
import pandas as pd
import joblib
import torch
import torch.nn as nn

from scipy.sparse import coo_matrix
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.model_selection import GroupShuffleSplit, StratifiedShuffleSplit
from sklearn.neighbors import NearestNeighbors

warnings.filterwarnings("ignore")

# ============================================================
# STEP 1: CONFIGURATION
# ============================================================

BASE_DIR = r"F:\code\emotion_expression"

FUSED_FILE = os.path.join(
    BASE_DIR,
    "Step4_Feature_Fusion",
    "Multimodal_Fused_Dataset.csv"
)

MASTER_FILE = os.path.join(
    BASE_DIR,
    "Step2_Preprocessed",
    "music_emotion_dataset_preprocessed.csv"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "Step6_to_Step9_Model"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

MODEL_FILE = os.path.join(
    OUTPUT_DIR,
    "ModBW_Att_GNNet_Model.pth"
)

PREDICTION_FILE = os.path.join(
    OUTPUT_DIR,
    "Emotion_Classification_Predictions.csv"
)

PARAMETER_FILE = os.path.join(
    OUTPUT_DIR,
    "ModBW_Optimized_Parameters.json"
)

PREPROCESSOR_FILE = os.path.join(
    OUTPUT_DIR,
    "Feature_Preprocessing_Objects.joblib"
)

RANDOM_STATE = 42
DEVICE = torch.device("cpu")

# Outer split: 70% training, 30% testing.
TEST_SIZE = 0.30

# Internal validation is taken only from the training set.
VALIDATION_SIZE = 0.20

# ModBW search configuration.
POPULATION_SIZE = 4
MODBW_ITERATIONS = 3
SEARCH_EPOCHS = 6

# Final training configuration.
FINAL_EPOCHS = 40

torch.manual_seed(RANDOM_STATE)
np.random.seed(RANDOM_STATE)

start_time = time.time()

print("=" * 75)
print("STEP 6 TO STEP 9: ModBW-Att-GNNet")
print("=" * 75)

if not os.path.isfile(FUSED_FILE):
    raise FileNotFoundError(FUSED_FILE)

# ============================================================
# STEP 2: LOAD FUSED DATASET
# ============================================================

print("\nLoading fused dataset...")

df = pd.read_csv(FUSED_FILE)

print("Dataset loaded successfully.")

TARGET = "Emotion_Class"

if TARGET not in df.columns:
    raise ValueError("Emotion_Class target column not found.")

# ============================================================
# STEP 3: IDENTIFY FEATURE GROUPS
# ============================================================

hog_columns = [
    c for c in df.columns
    if c.startswith("HOG_")
]

mfcc_columns = [
    c for c in df.columns
    if c.startswith("MFCC_")
]

physio_columns = [
    c for c in [
        "PPG_Related_Heart_Rate_BPM",
        "HRV_Score",
        "GSR_Normalized",
        "Respiration_Rate",
        "Body_Temperature_C",
        "Oxygen_Saturation",
        "Stress_Index",
        "Fatigue_Score"
    ]
    if c in df.columns
]

if not hog_columns or not mfcc_columns or not physio_columns:
    raise ValueError("One or more feature groups are missing.")

print("Image feature group identified.")
print("Audio feature group identified.")
print("Physiological feature group identified.")

# Convert features to numeric.
for col in hog_columns + mfcc_columns + physio_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

if df[
    hog_columns + mfcc_columns + physio_columns
].isnull().any().any():
    raise ValueError(
        "Missing or non-numeric feature values detected. "
        "Resolve them before training."
    )

# ============================================================
# STEP 4: TARGET ENCODING
# ============================================================

label_encoder = LabelEncoder()

y = label_encoder.fit_transform(
    df[TARGET].astype(str)
).astype(np.int64)

class_names = label_encoder.classes_.tolist()

num_classes = len(class_names)

print("\nEmotion target prepared.")

# ============================================================
# STEP 5: SUBJECT-AWARE TRAIN-TEST SPLITTING
# ============================================================

print("\nPreparing training and testing partitions...")

group_candidates = [
    "Subject_ID",
    "Participant_ID",
    "Performer_ID",
    "Musician_ID",
    "Artist_ID",
    "User_ID"
]

group_column = None

# Recover a subject identifier from the master dataset
# if one of the supported columns exists.
if os.path.isfile(MASTER_FILE):

    master = pd.read_csv(MASTER_FILE)

    available_groups = [
        c for c in group_candidates
        if c in master.columns
    ]

    if available_groups:

        group_column = available_groups[0]

        if "Performance_ID" not in master.columns:
            raise ValueError(
                "Performance_ID missing from master dataset."
            )

        group_lookup = master[
            ["Performance_ID", group_column]
        ].copy()

        if group_lookup["Performance_ID"].duplicated().any():
            raise ValueError(
                "Performance_ID is not unique in master dataset."
            )

        df = df.merge(
            group_lookup,
            on="Performance_ID",
            how="left",
            validate="one_to_one"
        )

        if df[group_column].isnull().any():
            raise ValueError(
                "Missing subject identifiers detected."
            )

        print("Subject identifier recovered:", group_column)

    del master
    gc.collect()

if group_column is not None:

    groups = df[group_column].astype(str).values

    outer_splitter = GroupShuffleSplit(
        n_splits=1,
        test_size=TEST_SIZE,
        random_state=RANDOM_STATE
    )

    train_idx, test_idx = next(
        outer_splitter.split(
            df,
            y,
            groups
        )
    )

    inner_splitter = GroupShuffleSplit(
        n_splits=1,
        test_size=VALIDATION_SIZE,
        random_state=RANDOM_STATE
    )

    inner_train_local, val_local = next(
        inner_splitter.split(
            train_idx,
            y[train_idx],
            groups[train_idx]
        )
    )

    inner_train_idx = train_idx[inner_train_local]
    val_idx = train_idx[val_local]

else:

    # Performance_ID is used as a record-level fallback.
    # It cannot establish subject-isolated evaluation.
    print(
        "WARNING: No subject identifier found. "
        "Using stratified record-level splitting."
    )

    outer_splitter = StratifiedShuffleSplit(
        n_splits=1,
        test_size=TEST_SIZE,
        random_state=RANDOM_STATE
    )

    train_idx, test_idx = next(
        outer_splitter.split(df, y)
    )

    inner_splitter = StratifiedShuffleSplit(
        n_splits=1,
        test_size=VALIDATION_SIZE,
        random_state=RANDOM_STATE
    )

    inner_train_local, val_local = next(
        inner_splitter.split(
            train_idx,
            y[train_idx]
        )
    )

    inner_train_idx = train_idx[inner_train_local]
    val_idx = train_idx[val_local]

# Check that every emotion class is represented in training.
if len(np.unique(y[train_idx])) != num_classes:
    raise ValueError(
        "The training partition does not contain all emotion classes. "
        "Review the split or subject distribution."
    )

print("Data partitioning completed.")

# ============================================================
# STEP 6: HETEROGENEOUS GRAPH CONSTRUCTION
# ============================================================

print("\nSTEP 6: HETEROGENEOUS GRAPH CONSTRUCTION")

# Fit scalers and PCA only on the outer training partition.
# Separate dimensionality reduction is used for each modality.

modality_settings = [
    ("HOG", hog_columns, 64),
    ("MFCC", mfcc_columns, 16),
    ("PHYSIO", physio_columns, 8)
]

X_modalities = []
preprocessing_objects = {}

for modality_name, columns, requested_components in modality_settings:

    X_modality = df[columns].to_numpy(dtype=np.float32)

    scaler = StandardScaler()

    X_scaled_train = scaler.fit_transform(
        X_modality[train_idx]
    )

    X_scaled_all = scaler.transform(X_modality)

    n_components = min(
        requested_components,
        X_scaled_train.shape[0] - 1,
        X_scaled_train.shape[1]
    )

    pca = PCA(
        n_components=n_components,
        svd_solver="randomized",
        random_state=RANDOM_STATE
    )

    pca.fit(X_scaled_train)

    X_reduced = pca.transform(
        X_scaled_all
    ).astype(np.float32)

    X_modalities.append(X_reduced)

    preprocessing_objects[modality_name] = {
        "columns": columns,
        "scaler": scaler,
        "pca": pca
    }

    del X_modality, X_scaled_train, X_scaled_all
    gc.collect()

# Fuse reduced modality representations.
X = np.concatenate(
    X_modalities,
    axis=1
).astype(np.float32)

del X_modalities
gc.collect()

print("Multimodal node representations prepared.")

# ------------------------------------------------------------
# HETEROGENEOUS KNN GRAPH
# ------------------------------------------------------------

def construct_graph(features, source_indices, k):

    n_samples = features.shape[0]

    source_indices = np.asarray(
        source_indices,
        dtype=np.int64
    )

    if len(source_indices) < 2:
        raise ValueError(
            "Not enough source nodes for graph construction."
        )

    k = min(int(k), len(source_indices) - 1)

    source_features = features[source_indices]

    # Build a neighbor index using source nodes only.
    source_nn = NearestNeighbors(
        n_neighbors=k + 1,
        metric="cosine",
        algorithm="brute",
        n_jobs=-1
    )

    source_nn.fit(source_features)

    source_distances, source_neighbors = (
        source_nn.kneighbors(source_features)
    )

    row_indices = []
    col_indices = []
    weights = []

    source_lookup = {
        global_id: local_id
        for local_id, global_id in enumerate(source_indices)
    }

    # Training/source node connections.
    for local_i, global_i in enumerate(source_indices):

        added = 0

        for distance, local_j in zip(
            source_distances[local_i],
            source_neighbors[local_i]
        ):

            global_j = source_indices[local_j]

            if global_i == global_j:
                continue

            similarity = max(
                0.0,
                1.0 - float(distance)
            )

            row_indices.append(global_i)
            col_indices.append(global_j)
            weights.append(similarity)

            added += 1

            if added >= k:
                break

        # Self connection.
        row_indices.append(global_i)
        col_indices.append(global_i)
        weights.append(1.0)

    # Validation/test nodes connect only to source nodes.
    source_mask = np.zeros(
        n_samples,
        dtype=bool
    )

    source_mask[source_indices] = True

    query_indices = np.flatnonzero(~source_mask)

    if len(query_indices) > 0:

        query_distances, query_neighbors = (
            source_nn.kneighbors(
                features[query_indices],
                n_neighbors=k
            )
        )

        for local_i, global_i in enumerate(query_indices):

            for distance, local_j in zip(
                query_distances[local_i],
                query_neighbors[local_i]
            ):

                global_j = source_indices[local_j]

                similarity = max(
                    0.0,
                    1.0 - float(distance)
                )

                row_indices.append(global_i)
                col_indices.append(global_j)
                weights.append(similarity)

            # Self connection for query node.
            row_indices.append(global_i)
            col_indices.append(global_i)
            weights.append(1.0)

    adjacency = coo_matrix(
        (
            np.asarray(weights, dtype=np.float32),
            (
                np.asarray(row_indices),
                np.asarray(col_indices)
            )
        ),
        shape=(n_samples, n_samples),
        dtype=np.float32
    ).tocsr()

    # Row-normalize adjacency.
    degree = np.asarray(
        adjacency.sum(axis=1)
    ).flatten()

    inverse_degree = np.zeros_like(
        degree,
        dtype=np.float32
    )

    inverse_degree[degree > 0] = (
        1.0 / degree[degree > 0]
    )

    normalized = (
        adjacency.multiply(
            inverse_degree[:, None]
        )
    ).tocoo()

    indices = torch.tensor(
        np.vstack([
            normalized.row,
            normalized.col
        ]),
        dtype=torch.long
    )

    values = torch.tensor(
        normalized.data,
        dtype=torch.float32
    )

    return torch.sparse_coo_tensor(
        indices,
        values,
        normalized.shape
    ).coalesce().to(DEVICE)


print("Heterogeneous graph construction function prepared.")

# ============================================================
# STEP 7: ModBW HYPERPARAMETER OPTIMIZATION
# ============================================================

print("\nSTEP 7: ModBW HYPERPARAMETER OPTIMIZATION")

# Parameter vector:
# [learning rate, dropout, hidden units, graph neighbors]

LOWER = np.array([
    0.001,
    0.15,
    64,
    5
], dtype=np.float64)

UPPER = np.array([
    0.010,
    0.50,
    128,
    15
], dtype=np.float64)

GOLDEN_RATIO = (np.sqrt(5.0) - 1.0) / 2.0


def decode_parameters(position):

    position = np.clip(
        position,
        0.0,
        1.0
    )

    learning_rate = (
        0.001
        + position[0] * 0.009
    )

    dropout = (
        0.15
        + position[1] * 0.35
    )

    hidden_units = int(
        round(
            64 + position[2] * 64
        )
    )

    # Keep hidden dimensions divisible by 8.
    hidden_units = max(
        64,
        min(
            128,
            int(round(hidden_units / 8) * 8)
        )
    )

    neighbors = int(
        round(
            5 + position[3] * 10
        )
    )

    return {
        "learning_rate": float(learning_rate),
        "dropout": float(dropout),
        "hidden_units": hidden_units,
        "neighbors": neighbors
    }


# ------------------------------------------------------------
# DOUBLE CHAOTIC INITIALIZATION
# ------------------------------------------------------------

def chaotic_population(population_size, dimensions):

    population = np.zeros(
        (population_size, dimensions),
        dtype=np.float64
    )

    x = 0.371
    z = 0.619

    for i in range(population_size):

        for j in range(dimensions):

            x = 4.0 * x * (1.0 - x)
            z = 4.0 * z * (1.0 - z)

            population[i, j] = (
                0.5 * x + 0.5 * z
            )

    return np.clip(
        population,
        0.0,
        1.0
    )


# ------------------------------------------------------------
# ATTENTION GRAPH NEURAL NETWORK
# ------------------------------------------------------------

class GraphAttentionBlock(nn.Module):

    def __init__(self, input_dim, output_dim, dropout):

        super().__init__()

        self.linear = nn.Linear(
            input_dim,
            output_dim
        )

        self.attention = nn.Linear(
            output_dim * 2,
            output_dim
        )

        self.dropout = nn.Dropout(dropout)

        self.norm = nn.LayerNorm(output_dim)

    def forward(self, x, adjacency):

        transformed = self.linear(x)

        neighbor_information = torch.sparse.mm(
            adjacency,
            transformed
        )

        combined = torch.cat(
            [
                transformed,
                neighbor_information
            ],
            dim=1
        )

        attention_weights = torch.sigmoid(
            self.attention(combined)
        )

        output = (
            attention_weights * transformed
            +
            (1.0 - attention_weights)
            * neighbor_information
        )

        output = torch.relu(output)

        output = self.norm(output)

        output = self.dropout(output)

        return output


class ModBW_Att_GNNet(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim,
        num_classes,
        dropout
    ):

        super().__init__()

        self.input_layer = nn.Linear(
            input_dim,
            hidden_dim
        )

        self.gat1 = GraphAttentionBlock(
            hidden_dim,
            hidden_dim,
            dropout
        )

        self.gat2 = GraphAttentionBlock(
            hidden_dim,
            hidden_dim,
            dropout
        )

        self.classifier = nn.Sequential(
            nn.Linear(
                hidden_dim,
                hidden_dim // 2
            ),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(
                hidden_dim // 2,
                num_classes
            )
        )

    def forward(self, x, adjacency):

        x = torch.relu(
            self.input_layer(x)
        )

        x = self.gat1(
            x,
            adjacency
        )

        x = self.gat2(
            x,
            adjacency
        )

        return self.classifier(x)


# ------------------------------------------------------------
# PREPARE TENSORS
# ------------------------------------------------------------

X_tensor = torch.tensor(
    X,
    dtype=torch.float32,
    device=DEVICE
)

y_tensor = torch.tensor(
    y,
    dtype=torch.long,
    device=DEVICE
)

train_tensor = torch.tensor(
    train_idx,
    dtype=torch.long,
    device=DEVICE
)

inner_train_tensor = torch.tensor(
    inner_train_idx,
    dtype=torch.long,
    device=DEVICE
)

val_tensor = torch.tensor(
    val_idx,
    dtype=torch.long,
    device=DEVICE
)

test_tensor = torch.tensor(
    test_idx,
    dtype=torch.long,
    device=DEVICE
)


# ------------------------------------------------------------
# CANDIDATE FITNESS EVALUATION
# ------------------------------------------------------------

adjacency_cache = {}


def get_adjacency(source_indices, k):

    key = (
        tuple(np.asarray(source_indices).tolist()),
        int(k)
    )

    if key not in adjacency_cache:

        adjacency_cache[key] = construct_graph(
            X,
            source_indices,
            k
        )

    return adjacency_cache[key]


def evaluate_candidate(position):

    params = decode_parameters(position)

    adjacency = get_adjacency(
        inner_train_idx,
        params["neighbors"]
    )

    model = ModBW_Att_GNNet(
        input_dim=X.shape[1],
        hidden_dim=params["hidden_units"],
        num_classes=num_classes,
        dropout=params["dropout"]
    ).to(DEVICE)

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=params["learning_rate"],
        weight_decay=1e-4
    )

    criterion = nn.CrossEntropyLoss()

    model.train()

    for epoch in range(SEARCH_EPOCHS):

        optimizer.zero_grad()

        outputs = model(
            X_tensor,
            adjacency
        )

        loss = criterion(
            outputs[inner_train_tensor],
            y_tensor[inner_train_tensor]
        )

        loss.backward()

        optimizer.step()

    # Validation loss is used internally for optimization.
    model.eval()

    with torch.no_grad():

        outputs = model(
            X_tensor,
            adjacency
        )

        validation_loss = criterion(
            outputs[val_tensor],
            y_tensor[val_tensor]
        ).item()

    del model, optimizer, adjacency
    gc.collect()

    return validation_loss


# ------------------------------------------------------------
# ModBW SEARCH WITH GOLDEN-SINE GUIDANCE
# ------------------------------------------------------------

population = chaotic_population(
    POPULATION_SIZE,
    len(LOWER)
)

fitness = np.array([
    evaluate_candidate(position)
    for position in population
])

best_index = np.argmin(fitness)

best_position = population[
    best_index
].copy()

best_fitness = fitness[
    best_index
]

for iteration in range(MODBW_ITERATIONS):

    updated_population = population.copy()

    for i in range(POPULATION_SIZE):

        current = population[i]

        random_value = np.random.rand()

        angle = 2.0 * np.pi * random_value

        golden_guidance = (
            GOLDEN_RATIO
            * np.sin(angle)
            * np.abs(
                GOLDEN_RATIO * best_position - current
            )
        )

        candidate = (
            current
            + golden_guidance
            + np.random.uniform(
                -0.10,
                0.10,
                size=len(LOWER)
            )
        )

        # Chaotic mutation.
        mutation = np.random.rand(
            len(LOWER)
        )

        mutation = (
            4.0 * mutation * (1.0 - mutation)
        )

        candidate = (
            0.90 * candidate
            + 0.10 * mutation
        )

        updated_population[i] = np.clip(
            candidate,
            0.0,
            1.0
        )

    updated_fitness = np.array([
        evaluate_candidate(position)
        for position in updated_population
    ])

    for i in range(POPULATION_SIZE):

        if updated_fitness[i] < fitness[i]:

            population[i] = updated_population[i]

            fitness[i] = updated_fitness[i]

    current_best = np.argmin(fitness)

    if fitness[current_best] < best_fitness:

        best_fitness = fitness[current_best]

        best_position = population[
            current_best
        ].copy()

    gc.collect()

best_params = decode_parameters(
    best_position
)

with open(
    PARAMETER_FILE,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        best_params,
        file,
        indent=4
    )

print("ModBW optimization completed.")

# ============================================================
# STEP 8: FINAL ATTENTION-GNN TRAINING
# ============================================================

print("\nSTEP 8: ATTENTION-GNN TRAINING")

# Rebuild graph with the complete outer training set.
# Test nodes receive messages from training nodes only.

final_adjacency = get_adjacency(
    train_idx,
    best_params["neighbors"]
)

final_model = ModBW_Att_GNNet(
    input_dim=X.shape[1],
    hidden_dim=best_params["hidden_units"],
    num_classes=num_classes,
    dropout=best_params["dropout"]
).to(DEVICE)

optimizer = torch.optim.Adam(
    final_model.parameters(),
    lr=best_params["learning_rate"],
    weight_decay=1e-4
)

criterion = nn.CrossEntropyLoss()

final_model.train()

for epoch in range(FINAL_EPOCHS):

    optimizer.zero_grad()

    outputs = final_model(
        X_tensor,
        final_adjacency
    )

    loss = criterion(
        outputs[train_tensor],
        y_tensor[train_tensor]
    )

    loss.backward()

    optimizer.step()

    # No training metrics are printed.

# ============================================================
# STEP 9: SOFTMAX EMOTION CLASSIFICATION
# ============================================================

print("\nSTEP 9: SOFTMAX EMOTION CLASSIFICATION")

final_model.eval()

with torch.no_grad():

    logits = final_model(
        X_tensor,
        final_adjacency
    )

    probabilities = torch.softmax(
        logits,
        dim=1
    )

    predicted_indices = torch.argmax(
        probabilities,
        dim=1
    ).cpu().numpy()

# Save predictions for the held-out test records.
actual_labels = label_encoder.inverse_transform(
    y[test_idx]
)

predicted_labels = label_encoder.inverse_transform(
    predicted_indices[test_idx]
)

prediction_df = pd.DataFrame({

    "Performance_ID": df.iloc[
        test_idx
    ]["Performance_ID"].values,

    "Actual_Emotion": actual_labels,

    "Predicted_Emotion": predicted_labels
})

for class_index, class_name in enumerate(class_names):

    prediction_df[
        f"Probability_{class_name}"
    ] = probabilities[
        test_tensor,
        class_index
    ].cpu().numpy()

prediction_df.to_csv(
    PREDICTION_FILE,
    index=False
)

# ============================================================
# SAVE TRAINED MODEL AND PREPROCESSING OBJECTS
# ============================================================

torch.save(
    {
        "model_state_dict": final_model.state_dict(),
        "input_dim": X.shape[1],
        "hidden_dim": best_params["hidden_units"],
        "num_classes": num_classes,
        "dropout": best_params["dropout"],
        "class_names": class_names,
        "feature_dimensions": {
            "HOG": len(hog_columns),
            "MFCC": len(mfcc_columns),
            "PHYSIO": len(physio_columns)
        }
    },
    MODEL_FILE
)

joblib.dump(
    {
        "preprocessing": preprocessing_objects,
        "label_encoder": label_encoder,
        "best_parameters": best_params,
        "feature_columns": {
            "HOG": hog_columns,
            "MFCC": mfcc_columns,
            "PHYSIO": physio_columns
        }
    },
    PREPROCESSOR_FILE
)

# ============================================================
# FINAL COMPLETION
# ============================================================

del X_tensor, y_tensor
del final_adjacency, adjacency_cache

print("\nTRAINING COMPLETED")

```

# 5-FOLD CROSS-VALIDATION

```python

# ============================================================
# 5-FOLD CROSS-VALIDATION - ModBW-Att-GNNet
# ============================================================

import time
import gc
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

print("\n" + "=" * 90)
print("5-FOLD CROSS-VALIDATION: ModBW-Att-GNNet")
print("=" * 90)

cv_start_time = time.time()

# ------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------

N_SPLITS = 5
CV_EPOCHS = 40

cv_results = []

skf = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=42
)

# ------------------------------------------------------------
# CROSS-VALIDATION
# ------------------------------------------------------------

for fold, (train_fold_idx, test_fold_idx) in enumerate(
    skf.split(df, y), start=1
):

    print(f"FOLD {fold}/{N_SPLITS}")

    fold_start = time.time()

    # --------------------------------------------------------
    # FOLD-SPECIFIC FEATURE SCALING AND PCA
    # --------------------------------------------------------

    fold_modalities = []

    fold_settings = [
        ("HOG", hog_columns, 64),
        ("MFCC", mfcc_columns, 16),
        ("PHYSIO", physio_columns, 8)
    ]

    for modality_name, columns, requested_components in fold_settings:

        modality_data = df[
            columns
        ].to_numpy(dtype=np.float32)

        fold_scaler = StandardScaler()

        X_train_scaled = fold_scaler.fit_transform(
            modality_data[train_fold_idx]
        )

        X_all_scaled = fold_scaler.transform(
            modality_data
        )

        n_components = min(
            requested_components,
            X_train_scaled.shape[0] - 1,
            X_train_scaled.shape[1]
        )

        fold_pca = PCA(
            n_components=n_components,
            svd_solver="randomized",
            random_state=42
        )

        fold_pca.fit(X_train_scaled)

        X_reduced = fold_pca.transform(
            X_all_scaled
        ).astype(np.float32)

        fold_modalities.append(X_reduced)

        del modality_data
        del X_train_scaled
        del X_all_scaled

    # Fuse the three modalities.
    fold_X = np.concatenate(
        fold_modalities,
        axis=1
    ).astype(np.float32)

    del fold_modalities
    gc.collect()

    # --------------------------------------------------------
    # BUILD FOLD-SPECIFIC GRAPH
    # --------------------------------------------------------

    fold_adjacency = construct_graph(
        fold_X,
        train_fold_idx,
        best_params["neighbors"]
    )

    fold_X_tensor = torch.tensor(
        fold_X,
        dtype=torch.float32,
        device=DEVICE
    )

    fold_y_tensor = torch.tensor(
        y,
        dtype=torch.long,
        device=DEVICE
    )

    fold_train_tensor = torch.tensor(
        train_fold_idx,
        dtype=torch.long,
        device=DEVICE
    )

    fold_test_tensor = torch.tensor(
        test_fold_idx,
        dtype=torch.long,
        device=DEVICE
    )

    # --------------------------------------------------------
    # INITIALIZE MODEL
    # --------------------------------------------------------

    fold_model = ModBW_Att_GNNet(
        input_dim=fold_X.shape[1],
        hidden_dim=best_params["hidden_units"],
        num_classes=num_classes,
        dropout=best_params["dropout"]
    ).to(DEVICE)

    fold_optimizer = torch.optim.Adam(
        fold_model.parameters(),
        lr=best_params["learning_rate"],
        weight_decay=1e-4
    )

    fold_criterion = nn.CrossEntropyLoss()

    # --------------------------------------------------------
    # TRAIN MODEL
    # --------------------------------------------------------

    fold_model.train()

    for epoch in range(CV_EPOCHS):

        fold_optimizer.zero_grad()

        fold_outputs = fold_model(
            fold_X_tensor,
            fold_adjacency
        )

        fold_loss = fold_criterion(
            fold_outputs[fold_train_tensor],
            fold_y_tensor[fold_train_tensor]
        )

        fold_loss.backward()

        fold_optimizer.step()

    # --------------------------------------------------------
    # TEST PREDICTIONS
    # --------------------------------------------------------

    fold_model.eval()

    with torch.no_grad():

        fold_outputs = fold_model(
            fold_X_tensor,
            fold_adjacency
        )

        fold_predictions = torch.argmax(
            fold_outputs[fold_test_tensor],
            dim=1
        ).cpu().numpy()

    y_actual = y[test_fold_idx]

    # --------------------------------------------------------
    # PERFORMANCE METRICS
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_actual,
        fold_predictions
    ) * 100

    precision = precision_score(
        y_actual,
        fold_predictions,
        average="macro",
        zero_division=0
    ) * 100

    recall = recall_score(
        y_actual,
        fold_predictions,
        average="macro",
        zero_division=0
    ) * 100

    f1 = f1_score(
        y_actual,
        fold_predictions,
        average="macro",
        zero_division=0
    ) * 100

    fold_time = time.time() - fold_start

    cv_results.append({
        "Fold": f"Fold {fold}",
        "Accuracy (%)": accuracy,
        "Precision (%)": precision,
        "Recall (%)": recall,
        "F1-Score (%)": f1,
        "Running Time (s)": fold_time
    })

    # Release fold-specific memory.
    del fold_model
    del fold_optimizer
    del fold_adjacency
    del fold_X
    del fold_X_tensor
    del fold_y_tensor
    del fold_train_tensor
    del fold_test_tensor
    del fold_outputs

    gc.collect()

# ============================================================
# FINAL 5-FOLD RESULTS
# ============================================================

cv_df = pd.DataFrame(cv_results)

# Calculate arithmetic mean for the five folds.
average_values = cv_df[
    [
        "Accuracy (%)",
        "Precision (%)",
        "Recall (%)",
        "F1-Score (%)"
    ]
].mean()

average_row = pd.DataFrame([{
    "Fold": "Average",
    "Accuracy (%)": average_values["Accuracy (%)"],
    "Precision (%)": average_values["Precision (%)"],
    "Recall (%)": average_values["Recall (%)"],
    "F1-Score (%)": average_values["F1-Score (%)"],
    "Running Time (s)": cv_df["Running Time (s)"].mean()
}])

final_cv_df = pd.concat(
    [cv_df, average_row],
    ignore_index=True
)

# ============================================================
# PRINT TABLE
# ============================================================

print("\n" + "=" * 100)
print("5-FOLD CROSS-VALIDATION RESULTS - ModBW-Att-GNNet")
print("=" * 100)

print(
    f"{'Folds':<12}"
    f"{'Accuracy (%)':>16}"
    f"{'Precision (%)':>16}"
    f"{'Recall (%)':>16}"
    f"{'F1-Score (%)':>16}"
)

print("-" * 100)

for _, row in final_cv_df.iterrows():

    print(
        f"{row['Fold']:<12}"
        f"{row['Accuracy (%)']:>16.2f}"
        f"{row['Precision (%)']:>16.2f}"
        f"{row['Recall (%)']:>16.2f}"
        f"{row['F1-Score (%)']:>16.2f}"
    )

print("-" * 100)

# ============================================================
# SAVE RESULTS
# ============================================================

CV_OUTPUT = os.path.join(
    OUTPUT_DIR,
    "ModBW_Att_GNNet_5Fold_CV_Results.csv"
)

final_cv_df.to_csv(
    CV_OUTPUT,
    index=False
)
```

```python

# =====================================================================
# ABLATION STUDY: ModBW-Att-GNNet
# Complete standalone implementation
# =====================================================================

import os
import gc
import json
import time
import random
import warnings

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
import torch.nn.functional as F

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

warnings.filterwarnings("ignore")

# =====================================================================
# STEP 1: CONFIGURATION
# =====================================================================

BASE_DIR = r"F:\code\emotion_expression"

DATA_FILE = os.path.join(
    BASE_DIR,
    "Step4_Feature_Fusion",
    "Multimodal_Fused_Dataset.csv"
)

MODEL_DIR = os.path.join(
    BASE_DIR,
    "Step6_to_Step9_Model"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "Step5_Ablation"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

PARAMETER_FILE = os.path.join(
    MODEL_DIR,
    "ModBW_Optimized_Parameters.json"
)

RESULT_FILE = os.path.join(
    OUTPUT_DIR,
    "ModBW_Att_GNNet_Ablation_Results.csv"
)

RANDOM_STATE = 42
TEST_SIZE = 0.30
EPOCHS = 2

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("=" * 90)
print("ABLATION STUDY: ModBW-Att-GNNet")
print("=" * 90)

print("Device:", DEVICE)

# =====================================================================
# STEP 2: REPRODUCIBILITY
# =====================================================================

random.seed(RANDOM_STATE)
np.random.seed(RANDOM_STATE)
torch.manual_seed(RANDOM_STATE)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(RANDOM_STATE)

# =====================================================================
# STEP 3: LOAD FUSED DATASET
# =====================================================================

print("\nLoading fused dataset...")

if not os.path.isfile(DATA_FILE):
    raise FileNotFoundError(DATA_FILE)

df = pd.read_csv(DATA_FILE)

print("Dataset shape:", df.shape)

TARGET = "Emotion_Class"

if TARGET not in df.columns:
    raise ValueError("Emotion_Class target column not found.")

# Identify feature groups
hog_columns = [
    c for c in df.columns
    if c.startswith("HOG_")
]

mfcc_columns = [
    c for c in df.columns
    if c.startswith("MFCC_")
]

physio_columns = [
    c for c in [
        "Heart_Rate_BPM",
        "HRV_Score",
        "Skin_Conductivity",
        "Respiration_Rate",
        "Body_Temperature_C",
        "Oxygen_Saturation",
        "Stress_Index",
        "Fatigue_Score"
    ]
    if c in df.columns
]

print("HOG features:", len(hog_columns))
print("MFCC features:", len(mfcc_columns))
print("Physiological features:", len(physio_columns))

if not hog_columns or not mfcc_columns or not physio_columns:
    raise ValueError(
        "Required HOG, MFCC or physiological features are missing."
    )

# =====================================================================
# STEP 4: TARGET ENCODING
# =====================================================================

encoder = LabelEncoder()

y = encoder.fit_transform(
    df[TARGET].astype(str)
)

num_classes = len(encoder.classes_)

print("\nEmotion classes:", list(encoder.classes_))
print("Number of classes:", num_classes)

# =====================================================================
# STEP 5: TRAIN-TEST SPLIT
# =====================================================================

all_indices = np.arange(len(df))

train_idx, test_idx = train_test_split(
    all_indices,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y
)

print("\nTraining samples:", len(train_idx))
print("Testing samples:", len(test_idx))

# =====================================================================
# STEP 6: LOAD MODBW OPTIMIZED PARAMETERS
# =====================================================================

if not os.path.isfile(PARAMETER_FILE):
    raise FileNotFoundError(
        "Optimized parameter file not found: "
        + PARAMETER_FILE
        + "\nRun the ModBW optimization step first."
    )

with open(PARAMETER_FILE, "r") as f:
    saved_params = json.load(f)

# Support common parameter key names.
def get_parameter(data, possible_names, default=None):

    if isinstance(data, dict):

        for key in possible_names:
            if key in data:
                return data[key]

        for value in data.values():
            if isinstance(value, dict):
                result = get_parameter(
                    value,
                    possible_names,
                    None
                )
                if result is not None:
                    return result

    return default


optimized_params = {
    "lr": get_parameter(
        saved_params,
        ["lr", "learning_rate"],
        None
    ),
    "dropout": get_parameter(
        saved_params,
        ["dropout", "dropout_rate"],
        None
    ),
    "hidden_dim": get_parameter(
        saved_params,
        ["hidden_dim", "hidden_units"],
        None
    ),
    "neighbors": get_parameter(
        saved_params,
        ["neighbors", "n_neighbors", "k"],
        None
    )
}

if any(v is None for v in optimized_params.values()):
    raise ValueError(
        "Could not identify all four optimized parameters in "
        "ModBW_Optimized_Parameters.json.\n"
        "Expected: lr, dropout, hidden_dim, neighbors.\n"
        "Inspect the JSON structure and update the parameter keys."
    )

optimized_params["lr"] = float(optimized_params["lr"])
optimized_params["dropout"] = float(optimized_params["dropout"])
optimized_params["hidden_dim"] = int(optimized_params["hidden_dim"])
optimized_params["neighbors"] = int(optimized_params["neighbors"])

print("\nLoaded ModBW optimized parameters:")
print(optimized_params)

# =====================================================================
# STEP 7: FEATURE PREPARATION
# =====================================================================

def prepare_features(
    use_hog=True,
    use_mfcc=True,
    use_physio=True
):

    selected_groups = []

    if use_hog:
        selected_groups.append((hog_columns, 64))

    if use_mfcc:
        selected_groups.append((mfcc_columns, 16))

    if use_physio:
        selected_groups.append((physio_columns, 8))

    if not selected_groups:
        raise ValueError("No feature group selected.")

    processed_groups = []

    for columns, requested_components in selected_groups:

        values = df[columns].apply(
            pd.to_numeric,
            errors="coerce"
        )

        values = values.replace(
            [np.inf, -np.inf],
            np.nan
        )

        # Imputation values are learned from training data only.
        train_values = values.iloc[train_idx]

        medians = train_values.median().fillna(0)

        values = values.fillna(medians).fillna(0)

        X_raw = values.to_numpy(dtype=np.float32)

        scaler = StandardScaler()

        scaler.fit(X_raw[train_idx])

        X_scaled = scaler.transform(X_raw).astype(np.float32)

        n_components = min(
            requested_components,
            X_scaled.shape[1],
            len(train_idx) - 1
        )

        if X_scaled.shape[1] > n_components and n_components > 0:

            pca = PCA(
                n_components=n_components,
                random_state=RANDOM_STATE
            )

            pca.fit(X_scaled[train_idx])

            X_group = pca.transform(
                X_scaled
            ).astype(np.float32)

        else:
            X_group = X_scaled

        processed_groups.append(X_group)

    X_combined = np.concatenate(
        processed_groups,
        axis=1
    )

    return X_combined

# =====================================================================
# STEP 8: GRAPH CONSTRUCTION
# =====================================================================

def construct_graph(X, source_indices, k):

    n_samples = X.shape[0]

    source_indices = np.asarray(
        source_indices,
        dtype=int
    )

    if len(source_indices) < 2:
        raise ValueError("Not enough training nodes for graph construction.")

    k = max(1, min(int(k), len(source_indices) - 1))

    source_features = X[source_indices]

    knn = NearestNeighbors(
        n_neighbors=k + 1,
        metric="euclidean",
        n_jobs=-1
    )

    knn.fit(source_features)

    edge_sources = []
    edge_targets = []

    # Every node connects only to training nodes.
    for target in range(n_samples):

        distances, neighbors = knn.kneighbors(
            X[target].reshape(1, -1),
            return_distance=True
        )

        neighbors = neighbors[0]

        neighbor_indices = source_indices[neighbors]

        # Exclude self-connection; add self-loops separately.
        neighbor_indices = neighbor_indices[
            neighbor_indices != target
        ][:k]

        for source in neighbor_indices:

            edge_sources.append(int(source))
            edge_targets.append(int(target))

    # Self-loops
    edge_sources.extend(range(n_samples))
    edge_targets.extend(range(n_samples))

    edge_index = torch.tensor(
        [edge_sources, edge_targets],
        dtype=torch.long
    )

    return edge_index

# =====================================================================
# STEP 9: GRAPH ATTENTION LAYER
# =====================================================================

class GraphAttentionLayer(nn.Module):

    def __init__(
        self,
        input_dim,
        output_dim,
        dropout=0.2
    ):

        super().__init__()

        self.linear = nn.Linear(
            input_dim,
            output_dim,
            bias=False
        )

        self.att_src = nn.Parameter(
            torch.empty(output_dim)
        )

        self.att_dst = nn.Parameter(
            torch.empty(output_dim)
        )

        self.bias = nn.Parameter(
            torch.zeros(output_dim)
        )

        self.dropout = dropout

        nn.init.xavier_uniform_(self.linear.weight)
        nn.init.xavier_uniform_(self.att_src.unsqueeze(0))
        nn.init.xavier_uniform_(self.att_dst.unsqueeze(0))

    def forward(self, x, edge_index):

        source = edge_index[0]
        target = edge_index[1]

        h = self.linear(x)

        source_score = (
            h[source] * self.att_src
        ).sum(dim=1)

        target_score = (
            h[target] * self.att_dst
        ).sum(dim=1)

        scores = F.leaky_relu(
            source_score + target_score,
            negative_slope=0.2
        )

        # Numerically stable softmax grouped by target node.
        max_scores = torch.full(
            (x.shape[0],),
            -torch.inf,
            device=x.device
        )

        max_scores.scatter_reduce_(
            0,
            target,
            scores,
            reduce="amax",
            include_self=True
        )

        exp_scores = torch.exp(
            scores - max_scores[target]
        )

        denominator = torch.zeros(
            x.shape[0],
            device=x.device
        )

        denominator.index_add_(
            0,
            target,
            exp_scores
        )

        attention = exp_scores / (
            denominator[target] + 1e-12
        )

        attention = F.dropout(
            attention,
            p=self.dropout,
            training=self.training
        )

        messages = (
            h[source] * attention.unsqueeze(1)
        )

        output = torch.zeros_like(h)

        output.index_add_(
            0,
            target,
            messages
        )

        return output + self.bias

# =====================================================================
# STEP 10: ModBW-Att-GNNet
# =====================================================================

class ModBW_Att_GNNet(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim,
        num_classes,
        dropout
    ):

        super().__init__()

        self.input_layer = nn.Linear(
            input_dim,
            hidden_dim
        )

        self.attention1 = GraphAttentionLayer(
            hidden_dim,
            hidden_dim,
            dropout
        )

        self.attention2 = GraphAttentionLayer(
            hidden_dim,
            hidden_dim,
            dropout
        )

        self.classifier = nn.Linear(
            hidden_dim,
            num_classes
        )

        self.dropout = dropout

    def forward(self, x, edge_index):

        x = F.relu(self.input_layer(x))

        x = F.dropout(
            x,
            p=self.dropout,
            training=self.training
        )

        x = F.elu(
            self.attention1(x, edge_index)
        )

        x = F.dropout(
            x,
            p=self.dropout,
            training=self.training
        )

        x = F.elu(
            self.attention2(x, edge_index)
        )

        return self.classifier(x)

# =====================================================================
# STEP 11: WITHOUT ATTENTION-BASED GNN
# =====================================================================

class Plain_GCN(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim,
        num_classes,
        dropout
    ):

        super().__init__()

        self.fc1 = nn.Linear(
            input_dim,
            hidden_dim
        )

        self.fc2 = nn.Linear(
            hidden_dim,
            hidden_dim
        )

        self.classifier = nn.Linear(
            hidden_dim,
            num_classes
        )

        self.dropout = dropout

    def aggregate(self, x, edge_index):

        source = edge_index[0]
        target = edge_index[1]

        output = torch.zeros_like(x)

        output.index_add_(
            0,
            target,
            x[source]
        )

        degree = torch.zeros(
            x.shape[0],
            device=x.device
        )

        degree.index_add_(
            0,
            target,
            torch.ones(
                len(target),
                device=x.device
            )
        )

        return output / degree.clamp(min=1).unsqueeze(1)

    def forward(self, x, edge_index):

        x = self.aggregate(x, edge_index)

        x = F.relu(self.fc1(x))

        x = F.dropout(
            x,
            p=self.dropout,
            training=self.training
        )

        x = self.aggregate(x, edge_index)

        x = F.relu(self.fc2(x))

        return self.classifier(x)

# =====================================================================
# STEP 12: TRAIN AND EVALUATE EACH VARIANT
# =====================================================================

def run_experiment(
    model_name,
    X_array,
    use_attention=True,
    use_modbw=True
):

    print("\n" + "-" * 80)
    print("Running:", model_name)
    print("-" * 80)

    start_time = time.time()

    # Use optimized or fixed parameters.
    if use_modbw:

        lr = optimized_params["lr"]
        dropout = optimized_params["dropout"]
        hidden_dim = optimized_params["hidden_dim"]
        neighbors = optimized_params["neighbors"]

    else:

        # Fixed baseline parameters without ModBW optimization.
        lr = 0.001
        dropout = 0.30
        hidden_dim = 64
        neighbors = 10

    X_tensor = torch.tensor(
        X_array,
        dtype=torch.float32,
        device=DEVICE
    )

    y_tensor = torch.tensor(
        y,
        dtype=torch.long,
        device=DEVICE
    )

    edge_index = construct_graph(
        X_array,
        train_idx,
        neighbors
    ).to(DEVICE)

    if use_attention:

        model = ModBW_Att_GNNet(
            input_dim=X_array.shape[1],
            hidden_dim=hidden_dim,
            num_classes=num_classes,
            dropout=dropout
        ).to(DEVICE)

    else:

        model = Plain_GCN(
            input_dim=X_array.shape[1],
            hidden_dim=hidden_dim,
            num_classes=num_classes,
            dropout=dropout
        ).to(DEVICE)

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=lr,
        weight_decay=1e-4
    )

    # Training
    model.train()

    for epoch in range(EPOCHS):

        optimizer.zero_grad()

        logits = model(
            X_tensor,
            edge_index
        )

        loss = F.cross_entropy(
            logits[train_idx],
            y_tensor[train_idx]
        )

        loss.backward()

        optimizer.step()

        if (epoch + 1) % 10 == 0:
            print(
                f"Epoch [{epoch + 1}/{EPOCHS}] "
                f"Loss: {loss.item():.4f}"
            )

    # Testing
    model.eval()

    with torch.no_grad():

        logits = model(
            X_tensor,
            edge_index
        )

        predictions = torch.argmax(
            logits[test_idx],
            dim=1
        ).cpu().numpy()

    actual = y[test_idx]

    accuracy = accuracy_score(
        actual,
        predictions
    )

    precision = precision_score(
        actual,
        predictions,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        actual,
        predictions,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        actual,
        predictions,
        average="macro",
        zero_division=0
    )

    elapsed = time.time() - start_time

    result = {
        "Model Variant": model_name,
        "Accuracy (%)": accuracy * 100,
        "Precision (%)": precision * 100,
        "Recall (%)": recall * 100,
        "F1-Score (%)": f1 * 100,
        "Training Time (s)": elapsed
    }


    del model, optimizer, logits
    del X_tensor, y_tensor, edge_index

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return result

# =====================================================================
# STEP 13: EXECUTE ABLATION EXPERIMENTS
# =====================================================================

ablation_results = []

# Experiment 1: Proposed ModBW-Att-GNNet
X_full = prepare_features(
    use_hog=True,
    use_mfcc=True,
    use_physio=True
)

ablation_results.append(
    run_experiment(
        "ModBW-Att-GNNet [Proposed]",
        X_full,
        use_attention=True,
        use_modbw=True
    )
)

del X_full
gc.collect()

# Experiment 2: Without ModBW Optimization
X_full = prepare_features(
    use_hog=True,
    use_mfcc=True,
    use_physio=True
)

ablation_results.append(
    run_experiment(
        "Without ModBW Optimization",
        X_full,
        use_attention=True,
        use_modbw=False
    )
)

del X_full
gc.collect()

# Experiment 3: Without Attention-Based GNN
X_full = prepare_features(
    use_hog=True,
    use_mfcc=True,
    use_physio=True
)

ablation_results.append(
    run_experiment(
        "Without Attention-Based GNN",
        X_full,
        use_attention=False,
        use_modbw=True
    )
)

del X_full
gc.collect()

# Experiment 4: Without MFCC Feature Extraction
X_no_mfcc = prepare_features(
    use_hog=True,
    use_mfcc=False,
    use_physio=True
)

ablation_results.append(
    run_experiment(
        "Without MFCC Feature Extraction",
        X_no_mfcc,
        use_attention=True,
        use_modbw=True
    )
)

del X_no_mfcc
gc.collect()

# Experiment 5: Without HOG Feature Extraction
X_no_hog = prepare_features(
    use_hog=False,
    use_mfcc=True,
    use_physio=True
)

ablation_results.append(
    run_experiment(
        "Without HOG Feature Extraction",
        X_no_hog,
        use_attention=True,
        use_modbw=True
    )
)

del X_no_hog
gc.collect()

# =====================================================================
# STEP 14: SAVE AND PRINT RESULTS
# =====================================================================

results_df = pd.DataFrame(ablation_results)

results_df.to_csv(
    RESULT_FILE,
    index=False
)

print("\n" + "=" * 110)
print("ABLATION STUDY: FINAL RESULTS")
print("=" * 110)

print(
    results_df[
        [
            "Model Variant",
            "Accuracy (%)",
            "Precision (%)",
            "Recall (%)",
            "F1-Score (%)"
        ]
    ].to_string(
        index=False,
        formatters={
            "Accuracy (%)": "{:.3f}".format,
            "Precision (%)": "{:.3f}".format,
            "Recall (%)": "{:.3f}".format,
            "F1-Score (%)": "{:.3f}".format
        }
    )
)

```

```python

# =====================================================================
# COMPARATIVE MODEL PERFORMANCE EVALUATION
# VGG-16 | 1D CNN-Inception-GRU | LSTM-AHP | CGF-Net | ModBW-Att-GNNet
# =====================================================================

import os
import gc
import json
import time
import random
import warnings

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import TensorDataset, DataLoader

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.neighbors import NearestNeighbors

from sklearn.metrics import (
    accuracy_score,
    recall_score,
    f1_score
)

warnings.filterwarnings("ignore")

# =====================================================================
# STEP 1: CONFIGURATION
# =====================================================================

BASE_DIR = r"F:\code\emotion_expression"

DATA_FILE = os.path.join(
    BASE_DIR,
    "Step4_Feature_Fusion",
    "Multimodal_Fused_Dataset.csv"
)

PARAMETER_FILE = os.path.join(
    BASE_DIR,
    "Step6_to_Step9_Model",
    "ModBW_Optimized_Parameters.json"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "Step10_Comparative_Analysis"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

RESULT_FILE = os.path.join(
    OUTPUT_DIR,
    "Comparative_Model_Performance.csv"
)

RANDOM_STATE = 42
TEST_SIZE = 0.30
EPOCHS = 1
BATCH_SIZE = 128

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

random.seed(RANDOM_STATE)
np.random.seed(RANDOM_STATE)
torch.manual_seed(RANDOM_STATE)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(RANDOM_STATE)

print("=" * 100)
print("COMPARATIVE MODEL PERFORMANCE EVALUATION")
print("=" * 100)

print("Device:", DEVICE)

# =====================================================================
# STEP 2: LOAD DATASET
# =====================================================================

print("\nLoading multimodal fused dataset...")

if not os.path.isfile(DATA_FILE):
    raise FileNotFoundError(
        "Dataset not found: " + DATA_FILE
    )

df = pd.read_csv(DATA_FILE)

TARGET = "Emotion_Class"

if TARGET not in df.columns:
    raise ValueError("Emotion_Class target column not found.")

# Feature groups
hog_columns = [
    c for c in df.columns
    if c.startswith("HOG_")
]

mfcc_columns = [
    c for c in df.columns
    if c.startswith("MFCC_")
]

physio_columns = [
    c for c in [
        "Heart_Rate_BPM",
        "HRV_Score",
        "Skin_Conductivity",
        "Respiration_Rate",
        "Body_Temperature_C",
        "Oxygen_Saturation",
        "Stress_Index",
        "Fatigue_Score"
    ]
    if c in df.columns
]

if not hog_columns:
    raise ValueError("HOG features are missing.")

if not mfcc_columns:
    raise ValueError("MFCC features are missing.")

if not physio_columns:
    raise ValueError("Physiological features are missing.")

# =====================================================================
# STEP 3: TARGET ENCODING
# =====================================================================

encoder = LabelEncoder()

y = encoder.fit_transform(
    df[TARGET].astype(str)
)

NUM_CLASSES = len(encoder.classes_)


# =====================================================================
# STEP 4: TRAIN-TEST SPLIT
# =====================================================================

all_indices = np.arange(len(df))

train_idx, test_idx = train_test_split(
    all_indices,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y
)

# =====================================================================
# STEP 5: FEATURE PREPROCESSING
# Scaler and PCA are fitted using training records only.
# =====================================================================

def process_feature_group(columns, requested_components):

    values = df[columns].apply(
        pd.to_numeric,
        errors="coerce"
    )

    values = values.replace(
        [np.inf, -np.inf],
        np.nan
    )

    train_medians = (
        values.iloc[train_idx]
        .median()
        .fillna(0)
    )

    values = values.fillna(train_medians).fillna(0)

    X = values.to_numpy(dtype=np.float32)

    scaler = StandardScaler()

    scaler.fit(X[train_idx])

    X_scaled = scaler.transform(X).astype(np.float32)

    n_components = min(
        requested_components,
        X_scaled.shape[1],
        len(train_idx) - 1
    )

    if X_scaled.shape[1] > n_components:

        pca = PCA(
            n_components=n_components,
            random_state=RANDOM_STATE
        )

        pca.fit(X_scaled[train_idx])

        X_scaled = pca.transform(
            X_scaled
        ).astype(np.float32)

    return X_scaled


print("\nPreparing multimodal features...")

X_hog = process_feature_group(
    hog_columns,
    64
)

X_mfcc = process_feature_group(
    mfcc_columns,
    16
)

# IMPORTANT: Your dataset has six physiological features.
# The number of features is determined dynamically.
X_physio = process_feature_group(
    physio_columns,
    8
)

PHYSIO_DIM = X_physio.shape[1]

X_combined = np.concatenate(
    [
        X_hog,
        X_mfcc,
        X_physio
    ],
    axis=1
).astype(np.float32)

INPUT_DIM = X_combined.shape[1]

# =====================================================================
# STEP 6: TENSOR CONVERSION AND DATA LOADER
# =====================================================================

X_HOG = torch.tensor(X_hog, dtype=torch.float32)
X_MFCC = torch.tensor(X_mfcc, dtype=torch.float32)
X_PHYSIO = torch.tensor(X_physio, dtype=torch.float32)
X_ALL = torch.tensor(X_combined, dtype=torch.float32)

Y = torch.tensor(y, dtype=torch.long)

train_dataset = TensorDataset(
    X_HOG[train_idx],
    X_MFCC[train_idx],
    X_PHYSIO[train_idx],
    X_ALL[train_idx],
    Y[train_idx]
)

test_dataset = TensorDataset(
    X_HOG[test_idx],
    X_MFCC[test_idx],
    X_PHYSIO[test_idx],
    X_ALL[test_idx],
    Y[test_idx]
)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False
)

# =====================================================================
# STEP 7: VGG-16 STYLE 1D CNN
# =====================================================================

class VGG16_1D(nn.Module):

    def __init__(self, input_dim, num_classes):

        super().__init__()

        self.features = nn.Sequential(

            nn.Conv1d(1, 64, 3, padding=1),
            nn.ReLU(),

            nn.Conv1d(64, 64, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(64, 128, 3, padding=1),
            nn.ReLU(),

            nn.Conv1d(128, 128, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(128, 256, 3, padding=1),
            nn.ReLU(),

            nn.Conv1d(256, 256, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(256, 512, 3, padding=1),
            nn.ReLU(),

            nn.Conv1d(512, 512, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(512, 512, 3, padding=1),
            nn.ReLU(),

            nn.Conv1d(512, 512, 3, padding=1),
            nn.ReLU(),

            nn.AdaptiveAvgPool1d(1)
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(512, 256),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):

        x = x.unsqueeze(1)

        x = self.features(x)

        return self.classifier(x)


# =====================================================================
# STEP 8: INCEPTION BLOCK
# =====================================================================

class InceptionBlock1D(nn.Module):

    def __init__(self, in_channels, out_channels):

        super().__init__()

        branch_channels = out_channels // 4

        self.branch1 = nn.Conv1d(
            in_channels,
            branch_channels,
            kernel_size=1
        )

        self.branch3 = nn.Conv1d(
            in_channels,
            branch_channels,
            kernel_size=3,
            padding=1
        )

        self.branch5 = nn.Conv1d(
            in_channels,
            branch_channels,
            kernel_size=5,
            padding=2
        )

        self.pool_branch = nn.Sequential(
            nn.MaxPool1d(
                kernel_size=3,
                stride=1,
                padding=1
            ),
            nn.Conv1d(
                in_channels,
                branch_channels,
                kernel_size=1
            )
        )

    def forward(self, x):

        b1 = F.relu(self.branch1(x))
        b3 = F.relu(self.branch3(x))
        b5 = F.relu(self.branch5(x))
        bp = F.relu(self.pool_branch(x))

        return torch.cat(
            [b1, b3, b5, bp],
            dim=1
        )


# =====================================================================
# STEP 9: 1D CNN WITH INCEPTION-GRU
# =====================================================================

class CNN_Inception_GRU(nn.Module):

    def __init__(self, input_dim, num_classes):

        super().__init__()

        self.initial = nn.Conv1d(
            1,
            64,
            kernel_size=3,
            padding=1
        )

        self.inception = InceptionBlock1D(
            64,
            128
        )

        self.gru = nn.GRU(
            input_size=128,
            hidden_size=128,
            num_layers=2,
            batch_first=True,
            dropout=0.2,
            bidirectional=True
        )

        self.classifier = nn.Sequential(
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128, num_classes)
        )

    def forward(self, x):

        x = x.unsqueeze(1)

        x = F.relu(self.initial(x))

        x = self.inception(x)

        x = x.transpose(1, 2)

        output, _ = self.gru(x)

        x = output[:, -1, :]

        return self.classifier(x)


# =====================================================================
# STEP 10: LSTM-BASED MULTIMODAL AHP FUSION
# =====================================================================

class LSTM_AHP_Fusion(nn.Module):

    def __init__(
        self,
        num_classes,
        physio_dim
    ):

        super().__init__()

        self.hog_projection = nn.Linear(
            X_hog.shape[1],
            64
        )

        self.mfcc_projection = nn.Linear(
            X_mfcc.shape[1],
            64
        )

        # Corrected: actual physiological input dimension.
        self.physio_projection = nn.Linear(
            physio_dim,
            64
        )

        self.hog_lstm = nn.LSTM(
            input_size=64,
            hidden_size=64,
            batch_first=True
        )

        self.mfcc_lstm = nn.LSTM(
            input_size=64,
            hidden_size=64,
            batch_first=True
        )

        self.physio_lstm = nn.LSTM(
            input_size=64,
            hidden_size=64,
            batch_first=True
        )

        # AHP pairwise comparison matrix.
        ahp_matrix = torch.tensor([
            [1.0, 3.0, 3.0],
            [1.0 / 3.0, 1.0, 2.0],
            [1.0 / 3.0, 1.0 / 2.0, 1.0]
        ])

        eigenvalues, eigenvectors = torch.linalg.eig(
            ahp_matrix
        )

        index = torch.argmax(
            eigenvalues.real
        )

        weights = eigenvectors[:, index].real.abs()

        weights = weights / weights.sum()

        self.register_buffer(
            "ahp_weights",
            weights.float()
        )

        self.classifier = nn.Sequential(
            nn.Linear(64, 128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128, num_classes)
        )

    def forward(
        self,
        hog,
        mfcc,
        physio,
        all_features
    ):

        h = F.relu(
            self.hog_projection(hog)
        ).unsqueeze(1)

        a = F.relu(
            self.mfcc_projection(mfcc)
        ).unsqueeze(1)

        p = F.relu(
            self.physio_projection(physio)
        ).unsqueeze(1)

        _, (h_state, _) = self.hog_lstm(h)

        _, (a_state, _) = self.mfcc_lstm(a)

        _, (p_state, _) = self.physio_lstm(p)

        h = h_state[-1]
        a = a_state[-1]
        p = p_state[-1]

        fused = (
            self.ahp_weights[0] * h
            + self.ahp_weights[1] * a
            + self.ahp_weights[2] * p
        )

        return self.classifier(fused)


# =====================================================================
# STEP 11: CGF-NET
# Cross-Gated Fusion Network
# =====================================================================

class CGF_Net(nn.Module):

    def __init__(
        self,
        num_classes,
        physio_dim
    ):

        super().__init__()

        self.hog_fc = nn.Linear(
            X_hog.shape[1],
            128
        )

        self.mfcc_fc = nn.Linear(
            X_mfcc.shape[1],
            128
        )

        self.physio_fc = nn.Linear(
            physio_dim,
            128
        )

        self.gate_hog = nn.Linear(
            256,
            128
        )

        self.gate_mfcc = nn.Linear(
            256,
            128
        )

        self.gate_physio = nn.Linear(
            256,
            128
        )

        self.fusion = nn.Sequential(
            nn.Linear(384, 256),
            nn.ReLU(),
            nn.Dropout(0.3),

            nn.Linear(256, 128),
            nn.ReLU(),

            nn.Linear(128, num_classes)
        )

    def forward(
        self,
        hog,
        mfcc,
        physio,
        all_features
    ):

        h = F.relu(self.hog_fc(hog))
        a = F.relu(self.mfcc_fc(mfcc))
        p = F.relu(self.physio_fc(physio))

        gate_h = torch.sigmoid(
            self.gate_hog(
                torch.cat([h, a], dim=1)
            )
        )

        gate_a = torch.sigmoid(
            self.gate_mfcc(
                torch.cat([a, p], dim=1)
            )
        )

        gate_p = torch.sigmoid(
            self.gate_physio(
                torch.cat([p, h], dim=1)
            )
        )

        h = h * gate_h
        a = a * gate_a
        p = p * gate_p

        fused = torch.cat(
            [h, a, p],
            dim=1
        )

        return self.fusion(fused)


# =====================================================================
# STEP 12: GRAPH ATTENTION LAYER
# =====================================================================

class GraphAttentionLayer(nn.Module):

    def __init__(
        self,
        input_dim,
        output_dim,
        dropout
    ):

        super().__init__()

        self.linear = nn.Linear(
            input_dim,
            output_dim,
            bias=False
        )

        self.att_src = nn.Parameter(
            torch.empty(output_dim)
        )

        self.att_dst = nn.Parameter(
            torch.empty(output_dim)
        )

        self.bias = nn.Parameter(
            torch.zeros(output_dim)
        )

        self.dropout = dropout

        nn.init.xavier_uniform_(
            self.linear.weight
        )

        nn.init.xavier_uniform_(
            self.att_src.unsqueeze(0)
        )

        nn.init.xavier_uniform_(
            self.att_dst.unsqueeze(0)
        )

    def forward(self, x, edge_index):

        source = edge_index[0]
        target = edge_index[1]

        h = self.linear(x)

        scores = F.leaky_relu(
            (h[source] * self.att_src).sum(dim=1)
            + (h[target] * self.att_dst).sum(dim=1),
            negative_slope=0.2
        )

        max_scores = torch.full(
            (x.shape[0],),
            -torch.inf,
            device=x.device
        )

        max_scores.scatter_reduce_(
            0,
            target,
            scores,
            reduce="amax",
            include_self=True
        )

        exp_scores = torch.exp(
            scores - max_scores[target]
        )

        denominator = torch.zeros(
            x.shape[0],
            device=x.device
        )

        denominator.index_add_(
            0,
            target,
            exp_scores
        )

        attention = exp_scores / (
            denominator[target] + 1e-12
        )

        attention = F.dropout(
            attention,
            p=self.dropout,
            training=self.training
        )

        messages = (
            h[source] * attention.unsqueeze(1)
        )

        output = torch.zeros_like(h)

        output.index_add_(
            0,
            target,
            messages
        )

        return output + self.bias


# =====================================================================
# STEP 13: ModBW-Att-GNNet
# =====================================================================

class ModBW_Att_GNNet(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim,
        num_classes,
        dropout
    ):

        super().__init__()

        self.input_layer = nn.Linear(
            input_dim,
            hidden_dim
        )

        self.attention1 = GraphAttentionLayer(
            hidden_dim,
            hidden_dim,
            dropout
        )

        self.attention2 = GraphAttentionLayer(
            hidden_dim,
            hidden_dim,
            dropout
        )

        self.classifier = nn.Linear(
            hidden_dim,
            num_classes
        )

        self.dropout = dropout

    def forward(self, x, edge_index):

        x = F.relu(
            self.input_layer(x)
        )

        x = F.dropout(
            x,
            p=self.dropout,
            training=self.training
        )

        x = F.elu(
            self.attention1(x, edge_index)
        )

        x = F.dropout(
            x,
            p=self.dropout,
            training=self.training
        )

        x = F.elu(
            self.attention2(x, edge_index)
        )

        return self.classifier(x)


# =====================================================================
# STEP 14: LOAD OPTIMIZED PARAMETERS
# =====================================================================

def get_nested_parameter(data, names):

    if isinstance(data, dict):

        for name in names:

            if name in data:
                return data[name]

        for value in data.values():

            if isinstance(value, dict):

                result = get_nested_parameter(
                    value,
                    names
                )

                if result is not None:
                    return result

    return None


def load_optimized_parameters():

    defaults = {
        "lr": 0.001,
        "dropout": 0.30,
        "hidden_dim": 128,
        "neighbors": 10
    }

    if not os.path.isfile(PARAMETER_FILE):

        print(
            "\nWARNING: Optimized parameter file not found. "
            "Using fixed fallback parameters."
        )

        return defaults

    with open(PARAMETER_FILE, "r") as f:
        data = json.load(f)

    params = {
        "lr": get_nested_parameter(
            data,
            ["lr", "learning_rate"]
        ),

        "dropout": get_nested_parameter(
            data,
            ["dropout", "dropout_rate"]
        ),

        "hidden_dim": get_nested_parameter(
            data,
            ["hidden_dim", "hidden_units"]
        ),

        "neighbors": get_nested_parameter(
            data,
            ["neighbors", "n_neighbors", "k"]
        )
    }

    for key in defaults:

        if params[key] is None:
            params[key] = defaults[key]

    params["lr"] = float(params["lr"])
    params["dropout"] = float(params["dropout"])
    params["hidden_dim"] = int(params["hidden_dim"])
    params["neighbors"] = int(params["neighbors"])

    return params


optimized_params = load_optimized_parameters()

print("\nModBW optimized parameters:")
print(optimized_params)


# =====================================================================
# STEP 15: GRAPH CONSTRUCTION
# =====================================================================

def build_graph(X, source_indices, k):

    source_indices = np.asarray(
        source_indices,
        dtype=int
    )

    k = max(
        1,
        min(int(k), len(source_indices) - 1)
    )

    knn = NearestNeighbors(
        n_neighbors=k + 1,
        metric="euclidean",
        n_jobs=-1
    )

    knn.fit(X[source_indices])

    sources = []
    targets = []

    for target in range(len(X)):

        _, neighbor_positions = knn.kneighbors(
            X[target].reshape(1, -1)
        )

        neighbor_indices = source_indices[
            neighbor_positions[0]
        ]

        neighbor_indices = neighbor_indices[
            neighbor_indices != target
        ][:k]

        for source in neighbor_indices:

            sources.append(int(source))
            targets.append(int(target))

    # Add self connections
    sources.extend(range(len(X)))
    targets.extend(range(len(X)))

    return torch.tensor(
        [sources, targets],
        dtype=torch.long,
        device=DEVICE
    )


# =====================================================================
# STEP 16: MODEL INITIALIZATION
# =====================================================================

def initialize_models():

    models = {}

    models["VGG-16"] = VGG16_1D(
        INPUT_DIM,
        NUM_CLASSES
    )

    models["1D CNN with Inception-GRU"] = CNN_Inception_GRU(
        INPUT_DIM,
        NUM_CLASSES
    )

    models["LSTM-based Multimodal AHP Fusion"] = LSTM_AHP_Fusion(
        NUM_CLASSES,
        PHYSIO_DIM
    )

    models["CGF-Net"] = CGF_Net(
        NUM_CLASSES,
        PHYSIO_DIM
    )

    models["ModBW-Att-GNNet [Proposed]"] = ModBW_Att_GNNet(
        INPUT_DIM,
        optimized_params["hidden_dim"],
        NUM_CLASSES,
        optimized_params["dropout"]
    )

    return models


# =====================================================================
# STEP 17: TRAIN AND EVALUATE
# =====================================================================

def train_and_evaluate(model_name, model):

    print("\n" + "-" * 90)
    print("Training:", model_name)
    print("-" * 90)

    start_time = time.perf_counter()

    model = model.to(DEVICE)

    # Use optimized parameters for proposed model.
    # Use the same fixed baseline settings for comparison models.
    if model_name == "ModBW-Att-GNNet [Proposed]":

        lr = optimized_params["lr"]
        neighbors = optimized_params["neighbors"]

    else:

        lr = 0.001
        neighbors = 10

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=lr,
        weight_decay=1e-4
    )

    criterion = nn.CrossEntropyLoss()

    is_graph_model = (
        model_name == "ModBW-Att-GNNet [Proposed]"
    )

    if is_graph_model:

        edge_index = build_graph(
            X_combined,
            train_idx,
            neighbors
        )

        X_device = X_ALL.to(DEVICE)
        Y_device = Y.to(DEVICE)

    else:

        edge_index = None

    # -------------------------------------------------
    # TRAINING
    # -------------------------------------------------

    model.train()

    for epoch in range(EPOCHS):

        epoch_loss = 0.0

        if is_graph_model:

            # Full-batch graph training: one optimizer update
            # per epoch, not one repeated update per mini-batch.
            optimizer.zero_grad()

            logits = model(
                X_device,
                edge_index
            )

            loss = criterion(
                logits[train_idx],
                Y_device[train_idx]
            )

            loss.backward()
            optimizer.step()

            epoch_loss = loss.item()

        else:

            for hog, mfcc, physio, all_features, labels in train_loader:

                hog = hog.to(DEVICE)
                mfcc = mfcc.to(DEVICE)
                physio = physio.to(DEVICE)
                all_features = all_features.to(DEVICE)
                labels = labels.to(DEVICE)

                optimizer.zero_grad()

                if model_name == "VGG-16":

                    logits = model(all_features)

                elif model_name == "1D CNN with Inception-GRU":

                    logits = model(all_features)

                elif model_name == "LSTM-based Multimodal AHP Fusion":

                    logits = model(
                        hog,
                        mfcc,
                        physio,
                        all_features
                    )

                elif model_name == "CGF-Net":

                    logits = model(
                        hog,
                        mfcc,
                        physio,
                        all_features
                    )

                loss = criterion(
                    logits,
                    labels
                )

                loss.backward()
                optimizer.step()

                epoch_loss += loss.item()

        if (epoch + 1) % 10 == 0:

            print(
                f"Epoch [{epoch + 1}/{EPOCHS}] "
                f"Loss: {epoch_loss:.4f}"
            )

    # -------------------------------------------------
    # TESTING
    # -------------------------------------------------

    model.eval()

    actual_all = []
    predicted_all = []

    with torch.no_grad():

        if is_graph_model:

            logits = model(
                X_device,
                edge_index
            )

            predictions = torch.argmax(
                logits[test_idx],
                dim=1
            ).cpu().numpy()

            actual = y[test_idx]

            actual_all.extend(
                actual.tolist()
            )

            predicted_all.extend(
                predictions.tolist()
            )

        else:

            for hog, mfcc, physio, all_features, labels in test_loader:

                hog = hog.to(DEVICE)
                mfcc = mfcc.to(DEVICE)
                physio = physio.to(DEVICE)
                all_features = all_features.to(DEVICE)

                if model_name == "VGG-16":

                    logits = model(all_features)

                elif model_name == "1D CNN with Inception-GRU":

                    logits = model(all_features)

                elif model_name == "LSTM-based Multimodal AHP Fusion":

                    logits = model(
                        hog,
                        mfcc,
                        physio,
                        all_features
                    )

                elif model_name == "CGF-Net":

                    logits = model(
                        hog,
                        mfcc,
                        physio,
                        all_features
                    )

                predictions = torch.argmax(
                    logits,
                    dim=1
                ).cpu().numpy()

                actual_all.extend(
                    labels.numpy().tolist()
                )

                predicted_all.extend(
                    predictions.tolist()
                )

    # -------------------------------------------------
    # PERFORMANCE METRICS
    # -------------------------------------------------

    accuracy = accuracy_score(
        actual_all,
        predicted_all
    )

    recall = recall_score(
        actual_all,
        predicted_all,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        actual_all,
        predicted_all,
        average="macro",
        zero_division=0
    )

    running_time = time.perf_counter() - start_time

    result = {
        "Methods": model_name,
        "Accuracy (%)": accuracy * 100,
        "Recall (%)": recall * 100,
        "F1-score (%)": f1 * 100,
        "Running Time (s)": running_time
    }

    del optimizer
    del logits

    if is_graph_model:

        del X_device
        del Y_device
        del edge_index

    del model

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return result


# =====================================================================
# STEP 18: RUN ALL FIVE MODELS
# =====================================================================

models = initialize_models()

results = []

for model_name, model in models.items():

    result = train_and_evaluate(
        model_name,
        model
    )

    results.append(result)

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()


# =====================================================================
# STEP 19: SAVE COMPARATIVE RESULTS
# =====================================================================

results_df = pd.DataFrame(results)

results_df.to_csv(
    RESULT_FILE,
    index=False
)

# =====================================================================
# STEP 20: PRINT FINAL COMPARISON TABLE
# =====================================================================

print("\n")
print("=" * 110)
print("COMPARATIVE MODEL PERFORMANCE RESULTS")
print("=" * 110)

print(
    results_df.to_string(
        index=False,
        formatters={
            "Accuracy (%)": "{:.3f}".format,
            "Recall (%)": "{:.3f}".format,
            "F1-score (%)": "{:.3f}".format,
            "Running Time (s)": "{:.3f}".format
        }
    )
)
```

```python

# =====================================================================
# STATISTICAL SIGNIFICANCE ANALYSIS
# ModBW-Att-GNNet vs CGF-Net
# Five-Fold Paired t-test
# =====================================================================

import os
import json
import gc
import time
import random
import warnings

import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import (
    accuracy_score,
    recall_score,
    f1_score
)

from scipy.stats import ttest_rel, t

warnings.filterwarnings("ignore")

# =====================================================================
# STEP 1: CONFIGURATION
# =====================================================================

BASE_DIR = r"F:\code\emotion_expression"

DATA_FILE = os.path.join(
    BASE_DIR,
    "Step4_Feature_Fusion",
    "Multimodal_Fused_Dataset.csv"
)

PARAMETER_FILE = os.path.join(
    BASE_DIR,
    "Step6_to_Step9_Model",
    "ModBW_Optimized_Parameters.json"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "Step11_Statistical_Analysis"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

FOLD_RESULT_FILE = os.path.join(
    OUTPUT_DIR,
    "FiveFold_Proposed_vs_CGF_Fold_Results.csv"
)

FINAL_RESULT_FILE = os.path.join(
    OUTPUT_DIR,
    "Statistical_Significance_Results.csv"
)

N_SPLITS = 5
EPOCHS = 40
BATCH_SIZE = 128
RANDOM_STATE = 42

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

random.seed(RANDOM_STATE)
np.random.seed(RANDOM_STATE)
torch.manual_seed(RANDOM_STATE)

print("=" * 100)
print("STATISTICAL SIGNIFICANCE ANALYSIS")
print("ModBW-Att-GNNet vs CGF-Net")
print("=" * 100)


# =====================================================================
# STEP 2: LOAD DATASET
# =====================================================================

df = pd.read_csv(DATA_FILE)

TARGET = "Emotion_Class"

hog_columns = [
    c for c in df.columns
    if c.startswith("HOG_")
]

mfcc_columns = [
    c for c in df.columns
    if c.startswith("MFCC_")
]

physio_columns = [
    c for c in [
        "Heart_Rate_BPM",
        "HRV_Score",
        "Skin_Conductivity",
        "Respiration_Rate",
        "Body_Temperature_C",
        "Oxygen_Saturation",
        "Stress_Index",
        "Fatigue_Score"
    ]
    if c in df.columns
]

from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

y = encoder.fit_transform(
    df[TARGET].astype(str)
)

NUM_CLASSES = len(encoder.classes_)


# =====================================================================
# STEP 3: LOAD OPTIMIZED PARAMETERS
# =====================================================================

with open(PARAMETER_FILE, "r") as f:
    parameter_data = json.load(f)


def find_parameter(data, names):

    if isinstance(data, dict):

        for name in names:
            if name in data:
                return data[name]

        for value in data.values():

            if isinstance(value, dict):

                result = find_parameter(
                    value,
                    names
                )

                if result is not None:
                    return result

    return None


params = {
    "lr": find_parameter(
        parameter_data,
        ["lr", "learning_rate"]
    ),

    "dropout": find_parameter(
        parameter_data,
        ["dropout", "dropout_rate"]
    ),

    "hidden_dim": find_parameter(
        parameter_data,
        ["hidden_dim", "hidden_units"]
    ),

    "neighbors": find_parameter(
        parameter_data,
        ["neighbors", "n_neighbors", "k"]
    )
}

defaults = {
    "lr": 0.001,
    "dropout": 0.30,
    "hidden_dim": 128,
    "neighbors": 10
}

for key in defaults:

    if params[key] is None:
        params[key] = defaults[key]

params["lr"] = float(params["lr"])
params["dropout"] = float(params["dropout"])
params["hidden_dim"] = int(params["hidden_dim"])
params["neighbors"] = int(params["neighbors"])


# =====================================================================
# STEP 4: FIVE-FOLD CROSS-VALIDATION
# =====================================================================

skf = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=RANDOM_STATE
)

fold_results = []

# =====================================================================
# STEP 5: FOLD-SPECIFIC FEATURE PREPARATION
# =====================================================================

def prepare_fold_features(train_indices, test_indices):

    feature_groups = []

    groups = [
        (hog_columns, 64),
        (mfcc_columns, 16),
        (physio_columns, 8)
    ]

    for columns, requested_components in groups:

        values = df[columns].apply(
            pd.to_numeric,
            errors="coerce"
        )

        values = values.replace(
            [np.inf, -np.inf],
            np.nan
        )

        medians = (
            values.iloc[train_indices]
            .median()
            .fillna(0)
        )

        values = values.fillna(medians).fillna(0)

        X_raw = values.to_numpy(dtype=np.float32)

        scaler = StandardScaler()

        scaler.fit(
            X_raw[train_indices]
        )

        X_scaled = scaler.transform(
            X_raw
        ).astype(np.float32)

        n_components = min(
            requested_components,
            X_scaled.shape[1],
            len(train_indices) - 1
        )

        if X_scaled.shape[1] > n_components:

            pca = PCA(
                n_components=n_components,
                random_state=RANDOM_STATE
            )

            pca.fit(
                X_scaled[train_indices]
            )

            X_scaled = pca.transform(
                X_scaled
            ).astype(np.float32)

        feature_groups.append(X_scaled)

    X_hog_fold = feature_groups[0]
    X_mfcc_fold = feature_groups[1]
    X_physio_fold = feature_groups[2]

    X_all_fold = np.concatenate(
        feature_groups,
        axis=1
    ).astype(np.float32)

    return (
        X_hog_fold,
        X_mfcc_fold,
        X_physio_fold,
        X_all_fold
    )


# =====================================================================
# STEP 6: TRAIN CGF-NET
# =====================================================================

def train_cgf(
    X_hog,
    X_mfcc,
    X_physio,
    train_indices,
    test_indices
):

    model = CGF_Net(
        num_classes=NUM_CLASSES,
        physio_dim=X_physio.shape[1]
    ).to(DEVICE)

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=0.001,
        weight_decay=1e-4
    )

    criterion = nn.CrossEntropyLoss()

    # Training tensors
    hog_train = torch.tensor(
        X_hog[train_indices],
        dtype=torch.float32,
        device=DEVICE
    )

    mfcc_train = torch.tensor(
        X_mfcc[train_indices],
        dtype=torch.float32,
        device=DEVICE
    )

    physio_train = torch.tensor(
        X_physio[train_indices],
        dtype=torch.float32,
        device=DEVICE
    )

    all_train = torch.tensor(
        np.concatenate(
            [
                X_hog[train_indices],
                X_mfcc[train_indices],
                X_physio[train_indices]
            ],
            axis=1
        ),
        dtype=torch.float32,
        device=DEVICE
    )

    y_train = torch.tensor(
        y[train_indices],
        dtype=torch.long,
        device=DEVICE
    )

    # Mini-batch training
    model.train()

    for epoch in range(EPOCHS):

        permutation = torch.randperm(
            len(train_indices),
            device=DEVICE
        )

        for start in range(
            0,
            len(train_indices),
            BATCH_SIZE
        ):

            batch = permutation[
                start:start + BATCH_SIZE
            ]

            optimizer.zero_grad()

            logits = model(
                hog_train[batch],
                mfcc_train[batch],
                physio_train[batch],
                all_train[batch]
            )

            loss = criterion(
                logits,
                y_train[batch]
            )

            loss.backward()
            optimizer.step()

    # Testing
    model.eval()

    hog_test = torch.tensor(
        X_hog[test_indices],
        dtype=torch.float32,
        device=DEVICE
    )

    mfcc_test = torch.tensor(
        X_mfcc[test_indices],
        dtype=torch.float32,
        device=DEVICE
    )

    physio_test = torch.tensor(
        X_physio[test_indices],
        dtype=torch.float32,
        device=DEVICE
    )

    all_test = torch.tensor(
        np.concatenate(
            [
                X_hog[test_indices],
                X_mfcc[test_indices],
                X_physio[test_indices]
            ],
            axis=1
        ),
        dtype=torch.float32,
        device=DEVICE
    )

    with torch.no_grad():

        logits = model(
            hog_test,
            mfcc_test,
            physio_test,
            all_test
        )

        predictions = torch.argmax(
            logits,
            dim=1
        ).cpu().numpy()

    del model, optimizer

    return predictions


# =====================================================================
# STEP 7: TRAIN ModBW-Att-GNNet
# =====================================================================

def train_proposed(
    X_all,
    train_indices,
    test_indices
):

    model = ModBW_Att_GNNet(
        input_dim=X_all.shape[1],
        hidden_dim=params["hidden_dim"],
        num_classes=NUM_CLASSES,
        dropout=params["dropout"]
    ).to(DEVICE)

    X_tensor = torch.tensor(
        X_all,
        dtype=torch.float32,
        device=DEVICE
    )

    y_tensor = torch.tensor(
        y,
        dtype=torch.long,
        device=DEVICE
    )

    edge_index = build_graph(
        X_all,
        train_indices,
        params["neighbors"]
    )

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=params["lr"],
        weight_decay=1e-4
    )

    criterion = nn.CrossEntropyLoss()

    model.train()

    for epoch in range(EPOCHS):

        optimizer.zero_grad()

        logits = model(
            X_tensor,
            edge_index
        )

        loss = criterion(
            logits[train_indices],
            y_tensor[train_indices]
        )

        loss.backward()
        optimizer.step()

    model.eval()

    with torch.no_grad():

        logits = model(
            X_tensor,
            edge_index
        )

        predictions = torch.argmax(
            logits[test_indices],
            dim=1
        ).cpu().numpy()

    del model, optimizer
    del X_tensor, y_tensor, edge_index

    return predictions


# =====================================================================
# STEP 8: RUN PAIRED FIVE-FOLD EXPERIMENTS
# =====================================================================

for fold, (train_indices, test_indices) in enumerate(
    skf.split(np.zeros(len(y)), y),
    start=1
):

    print(f"FOLD {fold}/{N_SPLITS}")
    fold_start = time.perf_counter()

    (
        X_hog_fold,
        X_mfcc_fold,
        X_physio_fold,
        X_all_fold
    ) = prepare_fold_features(
        train_indices,
        test_indices
    )

    # Same train/test indices for both methods.
    cgf_predictions = train_cgf(
        X_hog_fold,
        X_mfcc_fold,
        X_physio_fold,
        train_indices,
        test_indices
    )

    proposed_predictions = train_proposed(
        X_all_fold,
        train_indices,
        test_indices
    )

    actual = y[test_indices]

    # CGF-Net metrics
    cgf_accuracy = accuracy_score(
        actual,
        cgf_predictions
    )

    cgf_recall = recall_score(
        actual,
        cgf_predictions,
        average="macro",
        zero_division=0
    )

    cgf_f1 = f1_score(
        actual,
        cgf_predictions,
        average="macro",
        zero_division=0
    )

    # Proposed model metrics
    proposed_accuracy = accuracy_score(
        actual,
        proposed_predictions
    )

    proposed_recall = recall_score(
        actual,
        proposed_predictions,
        average="macro",
        zero_division=0
    )

    proposed_f1 = f1_score(
        actual,
        proposed_predictions,
        average="macro",
        zero_division=0
    )

    fold_time = time.perf_counter() - fold_start

    fold_results.append({

        "Fold": f"Fold {fold}",

        "Proposed_Accuracy": proposed_accuracy * 100,
        "CGF_Accuracy": cgf_accuracy * 100,

        "Proposed_Recall": proposed_recall * 100,
        "CGF_Recall": cgf_recall * 100,

        "Proposed_F1": proposed_f1 * 100,
        "CGF_F1": cgf_f1 * 100,

        "Fold_Running_Time": fold_time
    })


    del X_hog_fold
    del X_mfcc_fold
    del X_physio_fold
    del X_all_fold

    del cgf_predictions
    del proposed_predictions

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()


# =====================================================================
# STEP 9: SAVE FOLD-WISE RESULTS
# =====================================================================

fold_df = pd.DataFrame(fold_results)

fold_df.to_csv(
    FOLD_RESULT_FILE,
    index=False
)

print("\nFive-fold results saved:", FOLD_RESULT_FILE)


# =====================================================================
# STEP 10: PAIRED t-TEST AND 95% CONFIDENCE INTERVAL
# =====================================================================

def calculate_paired_statistics(
    proposed_values,
    cgf_values
):

    proposed_values = np.asarray(
        proposed_values,
        dtype=float
    )

    cgf_values = np.asarray(
        cgf_values,
        dtype=float
    )

    differences = (
        proposed_values - cgf_values
    )

    mean_proposed = np.mean(proposed_values)
    mean_cgf = np.mean(cgf_values)

    mean_difference = np.mean(differences)

    std_difference = np.std(
        differences,
        ddof=1
    )

    n = len(differences)
    dfree = n - 1

    standard_error = (
        std_difference / np.sqrt(n)
    )

    critical_value = t.ppf(
        0.975,
        dfree
    )

    lower_ci = (
        mean_difference
        - critical_value * standard_error
    )

    upper_ci = (
        mean_difference
        + critical_value * standard_error
    )

    if np.allclose(differences, 0):

        t_value = 0.0
        p_value = 1.0

    else:

        test_result = ttest_rel(
            proposed_values,
            cgf_values,
            alternative="two-sided"
        )

        t_value = float(test_result.statistic)
        p_value = float(test_result.pvalue)

    return {
        "ModBW-Att-GNNet [Proposed]": mean_proposed,
        "CGF-Net": mean_cgf,

        "Mean Difference (95% CI)": (
            f"{mean_difference:.3f} "
            f"({lower_ci:.3f}, {upper_ci:.3f})"
        ),

        "t-value (df = 4)": t_value,
        "p-value": p_value
    }


# =====================================================================
# STEP 11: CALCULATE ALL THREE METRICS
# =====================================================================

statistical_rows = []

metrics = [
    (
        "Accuracy",
        "Proposed_Accuracy",
        "CGF_Accuracy"
    ),
    (
        "Recall",
        "Proposed_Recall",
        "CGF_Recall"
    ),
    (
        "F1-score",
        "Proposed_F1",
        "CGF_F1"
    )
]

for metric_name, proposed_column, cgf_column in metrics:

    statistics = calculate_paired_statistics(
        fold_df[proposed_column].values,
        fold_df[cgf_column].values
    )

    statistics["Metrics"] = metric_name

    statistical_rows.append(statistics)


# =====================================================================
# STEP 12: FINAL TABLE
# =====================================================================

final_df = pd.DataFrame(
    statistical_rows
)

final_df = final_df[
    [
        "Metrics",
        "ModBW-Att-GNNet [Proposed]",
        "CGF-Net",
        "Mean Difference (95% CI)",
        "t-value (df = 4)",
        "p-value"
    ]
]

final_df.to_csv(
    FINAL_RESULT_FILE,
    index=False
)

# =====================================================================
# STEP 13: PRINT FINAL STATISTICAL TABLE
# =====================================================================

print("\n" + "=" * 150)
print("PAIRED STATISTICAL SIGNIFICANCE TEST")
print("ModBW-Att-GNNet vs CGF-Net")
print("=" * 150)

print(
    final_df.to_string(
        index=False,
        formatters={
            "ModBW-Att-GNNet [Proposed]": "{:.3f}".format,
            "CGF-Net": "{:.3f}".format,
            "t-value (df = 4)": "{:.4f}".format,
            "p-value": "{:.5f}".format
        }
    )
)
```

```python

# =====================================================================
# PERFORMANCE COMPARISON OF GNN MODELS
# GNN vs Att-GNN vs ModBW-Att-GNNet
# =====================================================================

import os
import gc
import json
import time
import random
import warnings

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
import torch.nn.functional as F

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.neighbors import NearestNeighbors

from sklearn.metrics import (
    accuracy_score,
    recall_score,
    f1_score,
    precision_score
)

warnings.filterwarnings("ignore")

# =====================================================================
# STEP 1: CONFIGURATION
# =====================================================================

BASE_DIR = r"F:\code\emotion_expression"

DATA_FILE = os.path.join(
    BASE_DIR,
    "Step4_Feature_Fusion",
    "Multimodal_Fused_Dataset.csv"
)

PARAMETER_FILE = os.path.join(
    BASE_DIR,
    "Step6_to_Step9_Model",
    "ModBW_Optimized_Parameters.json"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "Step12_GNN_Performance_Comparison"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

RESULT_FILE = os.path.join(
    OUTPUT_DIR,
    "GNN_Performance_Comparison.csv"
)

PREDICTION_FILE = os.path.join(
    OUTPUT_DIR,
    "GNN_Model_Predictions.csv"
)

RANDOM_STATE = 42
TEST_SIZE = 0.30
EPOCHS = 1

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

random.seed(RANDOM_STATE)
np.random.seed(RANDOM_STATE)
torch.manual_seed(RANDOM_STATE)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(RANDOM_STATE)

# =====================================================================
# STEP 2: LOAD DATASET
# =====================================================================

if not os.path.isfile(DATA_FILE):
    raise FileNotFoundError(DATA_FILE)

df = pd.read_csv(DATA_FILE)

TARGET = "Emotion_Class"

if TARGET not in df.columns:
    raise ValueError("Emotion_Class column not found.")

hog_columns = [
    c for c in df.columns
    if c.startswith("HOG_")
]

mfcc_columns = [
    c for c in df.columns
    if c.startswith("MFCC_")
]

physio_columns = [
    c for c in [
        "Heart_Rate_BPM",
        "HRV_Score",
        "Skin_Conductivity",
        "Respiration_Rate",
        "Body_Temperature_C",
        "Oxygen_Saturation",
        "Stress_Index",
        "Fatigue_Score"
    ]
    if c in df.columns
]

if not hog_columns or not mfcc_columns or not physio_columns:
    raise ValueError("One or more feature groups are missing.")

# =====================================================================
# STEP 3: TARGET ENCODING
# =====================================================================

encoder = LabelEncoder()

y = encoder.fit_transform(
    df[TARGET].astype(str)
)

NUM_CLASSES = len(encoder.classes_)


# =====================================================================
# STEP 4: TRAIN-TEST SPLIT
# =====================================================================

all_indices = np.arange(len(df))

train_idx, test_idx = train_test_split(
    all_indices,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y
)

# =====================================================================
# STEP 5: FEATURE PREPROCESSING
# =====================================================================

def prepare_feature_group(columns, components):

    values = df[columns].apply(
        pd.to_numeric,
        errors="coerce"
    )

    values = values.replace(
        [np.inf, -np.inf],
        np.nan
    )

    # Learn imputation values from training records only.
    medians = (
        values.iloc[train_idx]
        .median()
        .fillna(0)
    )

    values = values.fillna(medians).fillna(0)

    X = values.to_numpy(dtype=np.float32)

    scaler = StandardScaler()

    scaler.fit(X[train_idx])

    X_scaled = scaler.transform(X).astype(np.float32)

    n_components = min(
        components,
        X_scaled.shape[1],
        len(train_idx) - 1
    )

    if X_scaled.shape[1] > n_components:

        pca = PCA(
            n_components=n_components,
            random_state=RANDOM_STATE
        )

        pca.fit(X_scaled[train_idx])

        X_scaled = pca.transform(
            X_scaled
        ).astype(np.float32)

    return X_scaled


X_hog = prepare_feature_group(
    hog_columns,
    64
)

X_mfcc = prepare_feature_group(
    mfcc_columns,
    16
)

# Actual physiological feature dimension is determined automatically.
X_physio = prepare_feature_group(
    physio_columns,
    8
)

PHYSIO_DIM = X_physio.shape[1]

X_combined = np.concatenate(
    [X_hog, X_mfcc, X_physio],
    axis=1
).astype(np.float32)

# =====================================================================
# STEP 6: LOAD ModBW OPTIMIZED PARAMETERS
# =====================================================================

def find_parameter(data, names):

    if isinstance(data, dict):

        for name in names:
            if name in data:
                return data[name]

        for value in data.values():

            if isinstance(value, dict):

                result = find_parameter(
                    value,
                    names
                )

                if result is not None:
                    return result

    return None


def load_parameters():

    defaults = {
        "lr": 0.001,
        "dropout": 0.30,
        "hidden_dim": 128,
        "neighbors": 10
    }

    if not os.path.isfile(PARAMETER_FILE):

        print(
            " "
        )

        return defaults

    with open(PARAMETER_FILE, "r") as f:
        data = json.load(f)

    params = {
        "lr": find_parameter(
            data,
            ["lr", "learning_rate"]
        ),
        "dropout": find_parameter(
            data,
            ["dropout", "dropout_rate"]
        ),
        "hidden_dim": find_parameter(
            data,
            ["hidden_dim", "hidden_units"]
        ),
        "neighbors": find_parameter(
            data,
            ["neighbors", "n_neighbors", "k"]
        )
    }

    for key in defaults:

        if params[key] is None:
            params[key] = defaults[key]

    params["lr"] = float(params["lr"])
    params["dropout"] = float(params["dropout"])
    params["hidden_dim"] = int(params["hidden_dim"])
    params["neighbors"] = int(params["neighbors"])

    return params


optimized_params = load_parameters()


# =====================================================================
# STEP 7: GRAPH CONSTRUCTION
# =====================================================================

def construct_graph(X, source_indices, k):

    source_indices = np.asarray(
        source_indices,
        dtype=int
    )

    k = max(
        1,
        min(int(k), len(source_indices) - 1)
    )

    knn = NearestNeighbors(
        n_neighbors=k + 1,
        metric="euclidean",
        n_jobs=-1
    )

    knn.fit(X[source_indices])

    sources = []
    targets = []

    for target in range(len(X)):

        _, neighbors = knn.kneighbors(
            X[target].reshape(1, -1)
        )

        neighbor_indices = source_indices[
            neighbors[0]
        ]

        neighbor_indices = neighbor_indices[
            neighbor_indices != target
        ][:k]

        for source in neighbor_indices:

            sources.append(int(source))
            targets.append(int(target))

    # Self-connections
    sources.extend(range(len(X)))
    targets.extend(range(len(X)))

    return torch.tensor(
        [sources, targets],
        dtype=torch.long,
        device=DEVICE
    )


# =====================================================================
# STEP 8: GRAPH CONVOLUTION LAYER
# =====================================================================

class GraphConvolution(nn.Module):

    def __init__(self, input_dim, output_dim):

        super().__init__()

        self.linear = nn.Linear(
            input_dim,
            output_dim
        )

    def forward(self, x, edge_index):

        source = edge_index[0]
        target = edge_index[1]

        messages = self.linear(x[source])

        output = torch.zeros(
            (x.shape[0], messages.shape[1]),
            device=x.device
        )

        output.index_add_(
            0,
            target,
            messages
        )

        degree = torch.zeros(
            x.shape[0],
            device=x.device
        )

        degree.index_add_(
            0,
            target,
            torch.ones(
                len(target),
                device=x.device
            )
        )

        output = output / degree.clamp(
            min=1
        ).unsqueeze(1)

        return output


# =====================================================================
# STEP 9: STANDARD GNN MODEL
# =====================================================================

class GNN_Model(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim,
        num_classes,
        dropout
    ):

        super().__init__()

        self.gc1 = GraphConvolution(
            input_dim,
            hidden_dim
        )

        self.gc2 = GraphConvolution(
            hidden_dim,
            hidden_dim
        )

        self.classifier = nn.Linear(
            hidden_dim,
            num_classes
        )

        self.dropout = dropout

    def forward(self, x, edge_index):

        x = F.relu(
            self.gc1(x, edge_index)
        )

        x = F.dropout(
            x,
            p=self.dropout,
            training=self.training
        )

        x = F.relu(
            self.gc2(x, edge_index)
        )

        return self.classifier(x)


# =====================================================================
# STEP 10: GRAPH ATTENTION LAYER
# =====================================================================

class GraphAttentionLayer(nn.Module):

    def __init__(
        self,
        input_dim,
        output_dim,
        dropout
    ):

        super().__init__()

        self.linear = nn.Linear(
            input_dim,
            output_dim,
            bias=False
        )

        self.att_src = nn.Parameter(
            torch.empty(output_dim)
        )

        self.att_dst = nn.Parameter(
            torch.empty(output_dim)
        )

        self.bias = nn.Parameter(
            torch.zeros(output_dim)
        )

        self.dropout = dropout

        nn.init.xavier_uniform_(
            self.linear.weight
        )

        nn.init.xavier_uniform_(
            self.att_src.unsqueeze(0)
        )

        nn.init.xavier_uniform_(
            self.att_dst.unsqueeze(0)
        )

    def forward(self, x, edge_index):

        source = edge_index[0]
        target = edge_index[1]

        h = self.linear(x)

        scores = F.leaky_relu(
            (h[source] * self.att_src).sum(dim=1)
            + (h[target] * self.att_dst).sum(dim=1),
            negative_slope=0.2
        )

        max_scores = torch.full(
            (x.shape[0],),
            -torch.inf,
            device=x.device
        )

        max_scores.scatter_reduce_(
            0,
            target,
            scores,
            reduce="amax",
            include_self=True
        )

        exp_scores = torch.exp(
            scores - max_scores[target]
        )

        denominator = torch.zeros(
            x.shape[0],
            device=x.device
        )

        denominator.index_add_(
            0,
            target,
            exp_scores
        )

        attention = exp_scores / (
            denominator[target] + 1e-12
        )

        attention = F.dropout(
            attention,
            p=self.dropout,
            training=self.training
        )

        messages = (
            h[source] * attention.unsqueeze(1)
        )

        output = torch.zeros_like(h)

        output.index_add_(
            0,
            target,
            messages
        )

        return output + self.bias


# =====================================================================
# STEP 11: ATT-GNN MODEL
# =====================================================================

class Att_GNN_Model(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim,
        num_classes,
        dropout
    ):

        super().__init__()

        self.input_layer = nn.Linear(
            input_dim,
            hidden_dim
        )

        self.att1 = GraphAttentionLayer(
            hidden_dim,
            hidden_dim,
            dropout
        )

        self.att2 = GraphAttentionLayer(
            hidden_dim,
            hidden_dim,
            dropout
        )

        self.classifier = nn.Linear(
            hidden_dim,
            num_classes
        )

        self.dropout = dropout

    def forward(self, x, edge_index):

        x = F.relu(
            self.input_layer(x)
        )

        x = F.elu(
            self.att1(x, edge_index)
        )

        x = F.dropout(
            x,
            p=self.dropout,
            training=self.training
        )

        x = F.elu(
            self.att2(x, edge_index)
        )

        return self.classifier(x)


# =====================================================================
# STEP 12: PROPOSED ModBW-Att-GNNet
# =====================================================================

class ModBW_Att_GNNet(nn.Module):

    def __init__(
        self,
        branch_dims,
        hidden_dim,
        num_classes,
        dropout
    ):

        super().__init__()

        self.branch_layers = nn.ModuleList([
            nn.Linear(dim, hidden_dim)
            for dim in branch_dims
        ])

        # Learnable modality weights
        self.modality_weights = nn.Parameter(
            torch.zeros(len(branch_dims))
        )

        self.att1 = GraphAttentionLayer(
            hidden_dim,
            hidden_dim,
            dropout
        )

        self.att2 = GraphAttentionLayer(
            hidden_dim,
            hidden_dim,
            dropout
        )

        self.classifier = nn.Linear(
            hidden_dim,
            num_classes
        )

        self.dropout = dropout

    def forward(self, feature_groups, edge_index):

        representations = []

        for i, layer in enumerate(self.branch_layers):

            representation = F.relu(
                layer(feature_groups[i])
            )

            representations.append(
                representation
            )

        weights = torch.softmax(
            self.modality_weights,
            dim=0
        )

        fused = torch.zeros_like(
            representations[0]
        )

        for i, representation in enumerate(representations):

            fused = fused + (
                weights[i] * representation
            )

        fused = F.dropout(
            fused,
            p=self.dropout,
            training=self.training
        )

        fused = F.elu(
            self.att1(fused, edge_index)
        )

        fused = F.dropout(
            fused,
            p=self.dropout,
            training=self.training
        )

        fused = F.elu(
            self.att2(fused, edge_index)
        )

        return self.classifier(fused)


# =====================================================================
# STEP 13: TRAINING AND PREDICTION FUNCTION
# =====================================================================

def train_graph_model(
    model_name,
    X_groups,
    X_combined,
    params
):

    start_time = time.perf_counter()

    # Construct graph using training nodes.
    edge_index = construct_graph(
        X_combined,
        train_idx,
        params["neighbors"]
    )

    feature_tensors = [
        torch.tensor(
            group,
            dtype=torch.float32,
            device=DEVICE
        )
        for group in X_groups
    ]

    X_combined_tensor = torch.tensor(
        X_combined,
        dtype=torch.float32,
        device=DEVICE
    )

    y_tensor = torch.tensor(
        y,
        dtype=torch.long,
        device=DEVICE
    )

    hidden_dim = params["hidden_dim"]
    dropout = params["dropout"]

    # Initialize model
    if model_name == "GNN Model":

        model = GNN_Model(
            X_combined.shape[1],
            hidden_dim,
            NUM_CLASSES,
            dropout
        ).to(DEVICE)

        learning_rate = 0.001

    elif model_name == "Att-GNN Model":

        model = Att_GNN_Model(
            X_combined.shape[1],
            hidden_dim,
            NUM_CLASSES,
            dropout
        ).to(DEVICE)

        learning_rate = 0.001

    else:

        branch_dims = [
            group.shape[1]
            for group in X_groups
        ]

        model = ModBW_Att_GNNet(
            branch_dims,
            hidden_dim,
            NUM_CLASSES,
            dropout
        ).to(DEVICE)

        learning_rate = params["lr"]

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=learning_rate,
        weight_decay=1e-4
    )

    criterion = nn.CrossEntropyLoss()

    # -------------------------------------------------
    # TRAINING
    # -------------------------------------------------

    print("\nTraining:", model_name)

    model.train()

    for epoch in range(EPOCHS):

        optimizer.zero_grad()

        if model_name == "ModBW-Att-GNNet [Proposed Model]":

            logits = model(
                feature_tensors,
                edge_index
            )

        else:

            logits = model(
                X_combined_tensor,
                edge_index
            )

        loss = criterion(
            logits[train_idx],
            y_tensor[train_idx]
        )

        loss.backward()
        optimizer.step()

        if (epoch + 1) % 10 == 0:

            print(
                f"Epoch [{epoch + 1}/{EPOCHS}] "
            )

    # -------------------------------------------------
    # INFERENCE TIMING
    # -------------------------------------------------

    model.eval()

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()

    inference_start = time.perf_counter()

    with torch.no_grad():

        if model_name == "ModBW-Att-GNNet [Proposed Model]":

            logits = model(
                feature_tensors,
                edge_index
            )

        else:

            logits = model(
                X_combined_tensor,
                edge_index
            )

        predictions = torch.argmax(
            logits[test_idx],
            dim=1
        ).cpu().numpy()

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()

    inference_time = max(
        time.perf_counter() - inference_start,
        1e-9
    )

    throughput = len(test_idx) / inference_time

    actual = y[test_idx]

    # -------------------------------------------------
    # PERFORMANCE METRICS
    # -------------------------------------------------

    accuracy = accuracy_score(
        actual,
        predictions
    )

    recall = recall_score(
        actual,
        predictions,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        actual,
        predictions,
        average="macro",
        zero_division=0
    )

    weighted_f1 = f1_score(
        actual,
        predictions,
        average="weighted",
        zero_division=0
    )

    elapsed = time.perf_counter() - start_time

    result = {
        "Model": model_name,
        "Accuracy": accuracy * 100,
        "Recall": recall * 100,
        "F1": f1 * 100,
        "Weighted_F1": weighted_f1 * 100,
        "Inference_Time": inference_time,
        "Throughput": throughput,
        "Training_Time": elapsed
    }

    prediction_result = pd.DataFrame({
        "Actual_Emotion": encoder.inverse_transform(actual),
        "Predicted_Emotion": encoder.inverse_transform(predictions)
    })

    del model, optimizer, logits
    del feature_tensors, X_combined_tensor, y_tensor
    del edge_index

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return result, prediction_result


# =====================================================================
# STEP 14: EVALUATE THREE MODELS ON MULTIMODAL FEATURES
# =====================================================================

model_names = [
    "GNN Model",
    "Att-GNN Model",
    "ModBW-Att-GNNet [Proposed Model]"
]

comparison_results = []
prediction_results = {}

for model_name in model_names:

    # Use modality branches for the proposed model.
    # Baseline models use the concatenated representation.
    X_groups = [
        X_hog,
        X_mfcc,
        X_physio
    ]

    result, predictions = train_graph_model(
        model_name,
        X_groups,
        X_combined,
        optimized_params
    )

    comparison_results.append(result)

    prediction_results[model_name] = predictions


# =====================================================================
# STEP 15: MODALITY-SPECIFIC PHYSIOLOGICAL EVALUATION
# =====================================================================

print("\nEvaluating physiological signal analysis...")

physio_results = []

for model_name in model_names:

    result, _ = train_graph_model(
        model_name,
        [X_physio],
        X_physio,
        optimized_params
    )

    physio_results.append(result["Accuracy"])

# =====================================================================
# STEP 16: MODALITY-SPECIFIC ACOUSTIC EVALUATION
# =====================================================================

print("\nEvaluating acoustic emotion detection...")

acoustic_results = []

for model_name in model_names:

    result, _ = train_graph_model(
        model_name,
        [X_mfcc],
        X_mfcc,
        optimized_params
    )

    # Acoustic score is macro F1 on MFCC-only predictions.
    acoustic_results.append(result["F1"])

# =====================================================================
# STEP 17: CREATE SEVEN PERFORMANCE METRICS
# =====================================================================

final_results = []

for i, model_name in enumerate(model_names):

    result = comparison_results[i]

    # 1. Emotional expression prediction accuracy
    expression_accuracy = result["Accuracy"]

    # 2. Emotion recognition rate
    recognition_rate = result["Recall"]

    # 3. Physiological signal analysis accuracy
    physiological_accuracy = physio_results[i]

    # 4. Acoustic emotion detection score
    acoustic_score = acoustic_results[i]

    # 5. Emotion monitoring efficiency
    # Measured in test samples processed per second.
    monitoring_efficiency = result["Throughput"]

    # 6. Emotion classification rate
    # Measured using weighted F1-score.
    classification_rate = result["Weighted_F1"]

    # 7. Overall system reliability score
    # Custom composite of five percentage-based measures.
    reliability_score = np.mean([
        expression_accuracy,
        recognition_rate,
        physiological_accuracy,
        acoustic_score,
        classification_rate
    ])

    final_results.append({

        "Performance Metrics": "Emotional Expression Prediction Accuracy",
        "Model": model_name,
        "Value": expression_accuracy,
        "Unit": "%"

    })

    final_results.append({

        "Performance Metrics": "Emotion Recognition Rate",
        "Model": model_name,
        "Value": recognition_rate,
        "Unit": "%"

    })

    final_results.append({

        "Performance Metrics": "Physiological Signal Analysis Accuracy",
        "Model": model_name,
        "Value": physiological_accuracy,
        "Unit": "%"

    })

    final_results.append({

        "Performance Metrics": "Acoustic Emotion Detection Score",
        "Model": model_name,
        "Value": acoustic_score,
        "Unit": "%"

    })

    final_results.append({

        "Performance Metrics": "Emotion Monitoring Efficiency",
        "Model": model_name,
        "Value": monitoring_efficiency,
        "Unit": "samples/s"

    })

    final_results.append({

        "Performance Metrics": "Emotion Classification Rate",
        "Model": model_name,
        "Value": classification_rate,
        "Unit": "%"

    })

    final_results.append({

        "Performance Metrics": "Overall System Reliability Score",
        "Model": model_name,
        "Value": reliability_score,
        "Unit": "%"

    })

# =====================================================================
# STEP 18: CREATE FINAL COMPARISON TABLE
# =====================================================================

long_df = pd.DataFrame(final_results)

comparison_table = long_df.pivot(
    index="Performance Metrics",
    columns="Model",
    values="Value"
)

metric_order = [
    "Emotional Expression Prediction Accuracy",
    "Emotion Recognition Rate",
    "Physiological Signal Analysis Accuracy",
    "Acoustic Emotion Detection Score",
    "Emotion Monitoring Efficiency",
    "Emotion Classification Rate",
    "Overall System Reliability Score"
]

comparison_table = comparison_table.reindex(
    metric_order
)

comparison_table.to_csv(
    RESULT_FILE
)

# =====================================================================
# STEP 19: SAVE ADDITIONAL RESULTS
# =====================================================================

model_summary = pd.DataFrame(
    comparison_results
)

model_summary.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "GNN_Model_Overall_Metrics.csv"
    ),
    index=False
)

# Save predictions
prediction_frames = []

for model_name, prediction_df in prediction_results.items():

    temp = prediction_df.copy()
    temp["Model"] = model_name

    prediction_frames.append(temp)

all_predictions = pd.concat(
    prediction_frames,
    ignore_index=True
)

all_predictions.to_csv(
    PREDICTION_FILE,
    index=False
)

# =====================================================================
# STEP 20: PRINT FINAL TABLE
# =====================================================================

print("\n" + "=" * 145)
print("COMPARATIVE PERFORMANCE METRICS")
print("=" * 145)

print(
    comparison_table.to_string(
        float_format=lambda x: f"{x:.3f}"
    )
)
```
